# 2.1 配對交易

## 獲利機制的理論基礎

配對交易的獲利假設建立於**相對定價**而非絕對定價。

> 不主張能預測個股價格，僅主張兩檔具共同經濟驅動因子的證券，
> 其**相對**價格應維持在均衡關係附近；偏離且無基本面理由支持時，終將修正。

可追溯至「一價法則」（Law of One Price）在**近似替代品**上的弱化形式。

GGR (2006) 實證確認配對投資組合對 Fama-French 三因子的曝險極低 →
報酬難以由既有風險因子解釋 → 構成「異常報酬」的依據。

::: {.callout-important}

### 一個對本研究至關重要的區別

此機制**並未要求兩檔證券同屬一個產業**。

產業一致性是共同經濟衝擊的**充分而非必要條件**——
兩檔跨產業但受相同總體因子驅動的股票，理論上同樣可構成有效配對。

:::

## GGR (2006) 的基準設計

| 要素 | 設定 | 用意 |
| :--- | :--- | :--- |
| 期間分離 | 形成期 12 月 / 交易期 6 月 | 選對與交易不使用重疊資訊 |
| 價格正規化 | 累積總報酬指數（首日 = 1） | 使不同價位股票可比 |
| 距離準則 | 平方差總和（SSD）最小前 N 組 | 無參數、可完全複製 |
| 開倉門檻 | 偏離形成期 $\sigma$ 的 **2 倍** | 門檻具統計意涵 |
| 對沖比例 | 固定 **1**（等金額多空） | 不作回歸估計 |

**優點**：無參數、不需分布假設、可完全複製

**代價**：「歷史走勢相近」與「未來價差會回歸」之間**並無必然關係**

> 此缺口即後續共整合法與機器學習法試圖改善的對象。

## 報酬衰減：一致的實證發現

| 研究 | 貢獻 | 發現 |
| :--- | :--- | :--- |
| Do & Faff (2010) | 樣本延伸至 2009 | 獲利自 1990 年代末**顯著下降**，非特定次期間異常值所致 |
| Do & Faff (2012) | 納入實際成本（單邊 ~30bps） | 該成本下**多數策略淨報酬不再顯著** |
| Krauss (2017) | 五大類方法的文獻回顧 | **方法間差異往往小於樣本期間造成的差異** |

**Do & Faff (2010) 的一項觀察對本研究有直接意涵**：
衰減幅度在**流動性較高的大型股**上更為明顯。

::: {.callout-note}

### 本研究的定位

實證期間 2000–2025 年**完全落在衰減已被記錄之後**——這是刻意選擇。

問題不是「配對交易是否曾經獲利」，而是
「機器學習與深度學習能否在報酬已衰減的環境下**重建優勢**」。

故絕對獲利能力的檢定結果應理解為對衰減現象的**再確認**，
而非對兩大命題的直接答覆——**兩命題皆為方法間的相對比較**。

:::

# 2.2 配對篩選

## 方法演進：依假設強度排列

**距離法**（GGR, 2006）—— 假設最弱
僅要求歷史價格路徑接近，對價差的統計性質**無任何要求**。
Krauss (2017) 指出：距離最小者未必是價差最穩定者。
許鈞翔 (2025) 以 **DTW**（Sakoe & Chiba, 1978）容許時間軸非線性對齊。

**共整合法**（Engle & Granger, 1987；Vidyamurthy, 2004）
要求兩檔股票存在**定態的線性組合**——直接對「價差會回歸」此交易前提作檢定。
Krauss, Do & Huck (2016) 補入 **OU 半衰期**（回歸速度）
與 **Hurst 指數**（< 0.5 表均值回歸傾向）。

**Copula 法**（Liew & Wu, 2013）
放寬線性相依假設，可捕捉尾部與非對稱相依；代價是模型複雜度與分布敏感度。

> **孰優並無定論**：Huck & Afawubo (2015) 報告共整合較優；
> Rad, Low & Faff (2016) 在同一資料上發現三者差異有限、且隨樣本期間變動。

## 非監督式機器學習分群

**核心論點**：價格時間序列僅是公司特性的一個**投影**。
直接在特徵空間分群可避免偽相關，並發現價格路徑未顯示的經濟相似性。

| 研究 | 環境 | 方法 | 關鍵發現 |
| :--- | :--- | :--- | :--- |
| Avellaneda & Lee (2010) | 美股 | 報酬相關矩陣主成分（eigenportfolios） | 以**共同因子暴露**度量相似性 |
| Sarmento & Horta (2020) | 208 商品 ETF，5 分鐘 | OPTICS 密度分群 | 分群 **3.79** > 不分群 **3.58** > 依類別 **2.59** |
| **Han, He & Toh (2021)** | CRSP 全美股 1980–2020 | 48 動量 + **78 公司特徵**，Agglomerative | 年化 **24.8%**、Sharpe **2.69** |
| Hong & Hwang (2021) | 美股 | 基本面特徵距離（不分群） | 優於純價格法 |

**Han et al. 兩項對本研究至關重要的發現**

1. 獲利部分來自**跨產業**價差發散——長短邊常來自不同產業
2. 獲利**非**由小型股驅動——剔除 NYSE 規模後 20% 分位僅小幅降低，選股集中於規模上半部

## 一項未被充分討論的張力

各方法的核心差異在於**「相似性由什麼定義」**：

| 方法 | 相似性的定義 | 對價差性質的要求 |
| :--- | :--- | :--- |
| 距離法 | 歷史價格路徑接近 | 無 |
| 共整合法 | 存在定態線性組合 | **定態**（可加半衰期、Hurst） |
| Copula 法 | 聯合分布的相依結構 | 依 Copula 設定 |
| **機器學習分群** | **特徵空間的鄰近性** | **通常不作要求** |

::: {.callout-important}

### 機器學習分群與共整合篩選的定義並不一致

前者以**特徵相似性**為準，後者要求**價差平穩性**。
特徵相近的跨產業股票，未必具備使價差回歸的共同衝擊。

**Han et al. (2021) 的設計中並無共整合篩選**，
而多數以 GGR 為基礎的實作則保留之——兩者組合是否相容，文獻未見討論。

:::

此即本研究第四章受控消融的設計依據之一。

## 本研究與既有文獻的區隔（形成期）

**一、以 GICS 產業分類為對照組**
Sarmento & Horta (2020) 的「依類別分組」為 **5 個商品 ETF 類別**，經濟意涵較弱；
本研究以 **11 個 GICS 產業**為對照，構成強度高得多的先驗基準。

**二、單變因設計**
既有文獻多以各自最佳配置互相比較；
本研究固定特徵、排序、篩選與交易端，**僅變動分組方法**。

**三、對「分組」維度設置零點**
既有文獻比較的是不同的分組**方式**；
本研究另設「**完全不分組**」一格，使「限制搜尋空間」本身的價值可被評價。

# 2.3 交易策略

## 自固定門檻至學習式門檻選擇

**固定門檻**（GGR, 2006）
偏離 $2\sigma$ 進場、回歸至零平倉、期末強平。
優點：無參數估計、門檻具統計意涵。
**限制：門檻為外生給定且固定**——低波動期過於保守，高波動期過於激進。

**隨機控制與動態估計**
Elliott, van der Hoek & Malcolm (2005) 將價差建模為 **OU 過程**，
以 Kalman 濾波估計隱含狀態，推導最適邊界。
優勢：具明確最適性定義。代價：**真實動態偏離假設時最適性不成立**。
Do, Faff & Hamza (2006) 則主張對沖比例本身應隨時間變動。

**監督式學習**
Krauss, Do & Huck (2017) 以 DNN／GBT／隨機森林預測相對報酬——
惟其架構已近於**橫斷面選股**，而非傳統配對交易。

## 深度強化學習（DRL）路線

> Kim & Kim (2019)：以 DRL **同時最適化進場門檻與停損邊界**，
> 動作空間為門檻組合的離散集合，報酬函數為交易損益，
> 報告優於固定門檻的傳統規則。
>
> 朱羿璁 (2025)：結合共整合避險比率與混合獎勵設計，於台灣市場驗證類似架構。

**理論吸引力**：不需預測價格，僅需在給定狀態下**選擇門檻**——
問題結構與強化學習的「狀態—動作—報酬」框架天然契合（Sutton & Barto, 2018）。

::: {.callout-important}

### 本研究於此路線上的位置

本研究沿用 Kim & Kim (2019) 的**動作空間設計**（自離散的門檻組合中擇一），
但**不**沿用其學習範式。

在本研究的設定下，歷史配對期中全部 **9 個動作的報酬皆可精確反事實回算**，
故該學習問題為**全資訊監督回歸**而非部分回饋的強化學習（詳見 §3.4.4）。
本研究因此稱此交易端為 **DL-THR**，並另實作一個真正的部分回饋版本
（**RL-THR**：同一動作選單、同一狀態，僅改為只觀測所選動作的報酬 + ε-greedy）
作為受控對照，用以估計反事實標籤本身的價值。

:::

::: {.callout-warning}

### 此一區分具文獻上的意義

既有研究在同類設定下沿用「深度強化學習」之名時，
**未必檢視其問題結構是否真的具有探索與利用的取捨**。

若動作報酬可完整回算，強化學習的機制優勢即不成立——
觀察到的增益應歸因於**門檻選擇這個決策形式**，而非強化學習演算法。

:::

::: {.callout-tip}

### 交易期改良的方法論便利

可在**同一批配對**上施行 → 兩臂承受完全相同的市場風險、配對品質與期間
→ 構成天然的單變因對照。

此性質使命題 2 的檢定力遠高於命題 1（標準誤相差 1.89 倍，見 §4.1.3）。

:::

## 交易成本與本研究的區隔（交易期）

**Do & Faff (2012) 的方法論建議**
估計美股配對交易單邊成本約 **30 個基點**（含佣金與市場衝擊），
並主張報告 **break-even 成本**——使淨利歸零的成本水準——
以使結論不依賴單一成本假設。

### 本研究與既有文獻的區隔

**一、同一批配對上的單變因對照**
固定形成期產出，僅替換交易端。

**二、增益來源的分解**
既有文獻多僅報告學習式交易端是否較優。本研究將 DL-THR 的動作還原後，
分解為「**拒絕交易**」與「**調整門檻**」兩條管道，並對各管道設置對照。

**三、成本敏感度的量化**
依 Do & Faff (2012) 建議報告 break-even 成本，而非僅在單一假設下比較。

# 2.4 文獻缺口與貢獻定位

## 三項缺口

::: {.callout-important}

**其一：跨方法比較的推論基礎薄弱**
多數研究以各自參數網格的**最佳值**互相比較，未控制選擇偏誤；
且以參數組合為觀測單位的檢定，觀測間高度相關，**不滿足獨立性假設**。

**其二：形成期與交易期的改良常被混合施加**
當一項研究同時更換分群方法、篩選規則與交易端時，
其績效差異**無法歸因至任一單項**。

**其三：機器學習分群與共整合篩選的相容性未被檢驗**
兩者對「好配對」的定義不一致，而多數實作將兩者並用，**未討論其交互作用**。

:::

## 本研究的貢獻定位

| 缺口 | 本研究的填補方式 |
| :--- | :--- |
| 推論基礎 | 以**時間維度**的統計檢定取代參數網格檢定（逐日差分 + block bootstrap），並一律報全網格等權組合 |
| 歸因性 | 以**中性組裝器**實現單變因對照，每次僅動一個變因 |
| 交互作用 | 對「**分群 × 篩選**」設置受控消融，含「不分組」零點 |
