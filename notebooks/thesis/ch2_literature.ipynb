# 2.1 配對交易

## 獲利機制的理論基礎

配對交易的獲利假設建立於**相對定價**而非絕對定價。

> 不主張能預測個股價格，僅主張兩檔具共同經濟驅動因子的證券，
> 其**相對**價格應維持在均衡關係附近；偏離且無基本面理由支持時，終將修正。

可追溯至「一價法則」（Law of One Price）在**近似替代品**上的弱化形式。

**Gatev, Goetzmann & Rouwenhorst (2006)**（以下簡稱 GGR）
*Pairs trading: Performance of a relative-value arbitrage rule.*
**Review of Financial Studies, 19(3), 797–827.**

該研究實證確認配對投資組合對 Fama-French 三因子的曝險極低 →
報酬難以由既有風險因子解釋 → 構成「異常報酬」的依據。

::: {.callout-important}

### 一個對本研究至關重要的區別

此機制**並未要求兩檔證券同屬一個產業**。

產業一致性是共同經濟衝擊的**充分而非必要條件**——
兩檔跨產業但受相同總體因子驅動的股票，理論上同樣可構成有效配對。

:::


## GGR (2006) 的基準設計

| 要素 | 設定 | 用意 |
| :--- | :--- | :--- |
| 期間分離 | 形成期 12 月 / 交易期 6 月 | 選對與交易不使用重疊資訊 |
| 價格正規化 | 累積總報酬指數（首日 = 1） | 使不同價位股票可比 |
| 距離準則 | 平方差總和（SSD）最小前 N 組 | 無參數、可完全複製 |
| 開倉門檻 | 偏離形成期 $\sigma$ 的 **2 倍** | 門檻具統計意涵 |
| 對沖比例 | 固定 **1**（等金額多空） | 不作回歸估計 |

**優點**：無參數、不需分布假設、可完全複製

**代價**：「歷史走勢相近」與「未來價差會回歸」之間**並無必然關係**

> 此缺口即後續共整合法與機器學習法試圖改善的對象。


## 報酬衰減：一致的實證發現

| 研究 | 出處 | 貢獻 | 發現 |
| :--- | :--- | :--- | :--- |
| **Do & Faff (2010)** | *Does simple pairs trading still work?*<br/>**Financial Analysts Journal, 66(4), 83–95** | 樣本延伸至 2009 | 獲利自 1990 年代末**顯著下降**，非特定次期間異常值所致 |
| **Do & Faff (2012)** | *Are pairs trading profits robust to trading costs?*<br/>**Journal of Financial Research, 35(2), 261–287** | 納入實際成本（單邊 ~30bps） | 該成本下**多數策略淨報酬不再顯著** |
| **Krauss (2017)** | *Statistical arbitrage pairs trading strategies: Review and outlook.*<br/>**Journal of Economic Surveys, 31(2), 513–545**<br/>（工作論文版：IWQW Discussion Paper No. 09/2015） | 五大類方法的文獻回顧 | **方法間差異往往小於樣本期間造成的差異** |
| **Jacobs & Weber (2015)** | *On the determinants of pairs trading profitability.*<br/>**Journal of Financial Markets, 23, 75–97** | 獲利來源的橫斷面分析 | 獲利與**投資人注意力**及資訊擴散速度相關 |

**Do & Faff (2010) 的一項觀察對本研究有直接意涵**：
衰減幅度在**流動性較高的大型股**上更為明顯。

::: {.callout-note}

### 本研究的定位

實證期間 2000–2025 年**完全落在衰減已被記錄之後**——這是刻意選擇。
且本研究母體為 S&P 500 大型股，恰為 Do & Faff 指出衰減最嚴重之處。

問題不是「配對交易是否曾經獲利」，而是
「機器學習與門檻選擇能否在報酬已衰減的環境下**重建優勢**」。

故絕對獲利能力的檢定結果應理解為對衰減現象的**再確認**，
而非對兩大命題的直接答覆——**兩命題皆為方法間的相對比較**。

:::


# 2.2 配對篩選

## 方法演進：依假設強度排列

**距離法**——假設最弱
GGR (2006) 以正規化價格路徑的平方差總和（SSD）選對，
僅要求歷史價格路徑接近，對價差的統計性質**無任何要求**。
Krauss (2017) 指出：距離最小者未必是價差最穩定者。
許鈞翔 (2025)《最小距離法結合動態時間校正之配對交易研究》改以 **DTW**
（**Sakoe & Chiba, 1978**, *IEEE Transactions on Acoustics, Speech, and Signal
Processing, 26(1), 43–49*）容許時間軸的非線性對齊。

**共整合法**——直接檢定交易前提
**Engle & Granger (1987)** *Co-integration and error correction: Representation,
estimation, and testing.* **Econometrica, 55(2), 251–276.**
要求兩檔股票存在**定態的線性組合**。
**Vidyamurthy (2004)** *Pairs Trading: Quantitative Methods and Analysis.* Wiley.
將其系統性引入配對交易。
實務上另加兩道輔助濾網（見 Krauss, 2017 之回顧）：
**OU 半衰期**（價差回歸速度）與 **Hurst 指數**（$H<0.5$ 表均值回歸傾向）。

**Copula 法**——放寬線性相依
**Liew & Wu (2013)** *Pairs trading: A copula approach.*
**Journal of Derivatives & Hedge Funds, 19(1), 12–30.**
可捕捉尾部與非對稱相依；代價是模型複雜度與分布敏感度。

::: {.callout-warning}

### 孰優並無定論

| :--- | :--- | :--- |
| Huck & Afawubo (2015) | *Pairs trading and selection methods: Is cointegration superior?*<br/>**Applied Economics, 47(6), 599–613** | 共整合法較優 |
| Rad, Low & Faff (2016) | *The profitability of pairs trading strategies: Distance, cointegration, and copula methods.*<br/>**Quantitative Finance, 16(10), 1541–1558** | 三者差異有限，且**隨樣本期間變動** |

## 非監督式機器學習分群

**核心論點**：價格時間序列僅是公司特性的一個**投影**。
直接在特徵空間分群可避免偽相關，並發現價格路徑未顯示的經濟相似性。

| 研究 | 完整出處 | 環境與方法 | 關鍵發現 |
| :--- | :--- | :--- | :--- |
| **Avellaneda & Lee (2010)** | *Statistical arbitrage in the U.S. equities market.* **Quantitative Finance, 10(7), 761–782** | 美股；報酬相關矩陣主成分（eigenportfolios） | 以**共同因子暴露**度量相似性 |
| **Sarmento (2019)** | *Enhancing a Pairs Trading strategy with the application of Machine Learning.* **MSc thesis, Instituto Superior Técnico**（期刊版：Sarmento & Horta, 2020, *Expert Systems with Applications*, 158, 113490） | 208 檔商品型 ETF，5 分鐘，2009–2018；OPTICS 密度分群 | Sharpe：分群 **3.79** > 不分群 **3.58** > 依類別 **2.59** |
| **Han, He & Toh (2021)** | *Pairs trading via unsupervised learning.* **SSRN Working Paper No. 3860998**（未經同儕審查） | CRSP 全美股 1980–2020，月頻；48 動量因子 + 78 公司特徵，Agglomerative | 年化 **24.8%**、Sharpe **2.69**；扣 20bp 成本後 **15.9%**、Sharpe **1.73** |
| **Hong & Hwang** | *In search of pairs using firm fundamentals: Is pairs trading profitable?* 工作論文，Sungkyunkwan University | 美股；基本面特徵距離（**不分群**） | 特徵相似度較高的配對組合表現較佳 |

::: {.callout-important}

### Han et al. 的發現須依原文精確陳述——方向常被誤讀

原文（p.4）：

> *Clusters are homogeneous in terms of industry, consisting of only one or few
> sectors.* We also find that the long positions and short positions are often
> from **different sectors**, which implies that the profit does not only result
> from the divergence of the stocks in the same industry but also from the
> divergence **between** sectors.

**分群結果本身是產業同質的**（一個群多半只涵蓋一或少數幾個產業）。
獲利之所以含跨產業成分，是因為**同一群內的多、空兩腿常落在不同產業**，
**不是**因為分群跨越了產業界線。

→ 其主張是「**公司特徵**改善配對辨識」，而非「分群應該跨產業」。
   §2.2 末與第四章的消融設計須據此讀取。

**第二項發現**：剔除 NYSE 規模後 20% 分位僅使獲利小幅下降
→ 獲利**非**由小型股驅動，策略具規模承載力。

:::

::: {.callout-warning}

### Sarmento 的三臂比較有一項不利於分群的但書

原文（p.87）指出：若以 **ROI** 為準，**「完全不分群」整體反而較高**；
OPTICS 僅在**風險調整後**（Sharpe、獲利配對占比 86% vs 79%）勝出。

分群優於不分群的證據，並不如單看 Sharpe 那樣明確。

:::


## 一項未被充分討論的張力

各方法的核心差異在於**「相似性由什麼定義」**：

| 方法 | 相似性的定義 | 對價差性質的要求 |
| :--- | :--- | :--- |
| 距離法（GGR, 2006） | 歷史價格路徑接近 | 無 |
| 共整合法（Engle & Granger, 1987） | 存在定態線性組合 | **定態**（可加半衰期、Hurst） |
| Copula 法（Liew & Wu, 2013） | 聯合分布的相依結構 | 依 Copula 設定 |
| **機器學習分群**（Han et al., 2021） | **特徵空間的鄰近性** | **通常不作要求** |

::: {.callout-important}

### 機器學習分群與共整合篩選的定義並不一致

前者以**特徵相似性**為準，後者要求**價差平穩性**。
特徵相近的兩檔股票，未必具備使價差回歸的共同衝擊。

**Han, He & Toh (2021) 的設計中並無共整合篩選**，
而多數以 GGR 為基礎的實作則保留之——兩者組合是否相容，文獻未見討論。

:::

此即本研究第四章受控消融的設計依據之一。


## 本研究與既有文獻的區隔（形成期）

**一、以 GICS 產業分類為對照組**
Sarmento (2019) 的「依類別分組」為 **5 個商品 ETF 類別**，經濟意涵較弱；
本研究以 **11 個 GICS 產業**為對照，構成強度高得多的先驗基準。

**二、單變因設計**
既有文獻多以各自最佳配置互相比較；
本研究固定特徵、排序、篩選與交易端，**僅變動分組方法**。

**三、對「分組」維度設置零點**
既有文獻比較的是不同的分組**方式**；
本研究另設「**完全不分組**」一格，使「限制搜尋空間」本身的價值可被評價。
此設計呼應 Sarmento (2019) 的但書——該研究中不分群的 ROI 反而較高。

**四、母體與特徵條件的落差須事先聲明**
Han et al. (2021) 為 CRSP 全美股（數千檔）× 126 維特徵；
本研究為 S&P 500（每期約 600 檔）× SEC XBRL 可得子集。
**本研究不是該文的重製**，而是在受限資料條件下檢驗其方法主張是否仍成立。


# 2.3 交易策略

## 自固定門檻至學習式門檻選擇

**固定門檻**（GGR, 2006）
偏離形成期標準差 $2\sigma$ 進場、回歸至零平倉、期末強平。
優點：無參數估計、門檻具統計意涵。
**限制：門檻為外生給定且固定**——低波動期過於保守，高波動期過於激進。

**隨機控制與動態估計**

> **Elliott, Van Der Hoek & Malcolm (2005).** *Pairs trading.*
> **Quantitative Finance, 5(3), 271–276.**
> 將價差建模為 **OU 過程**，以 Kalman 濾波估計隱含狀態，推導最適進出場邊界。
>
> **Do, Faff & Hamza (2006).** *A new approach to modeling and estimation for
> pairs trading.* 主張**對沖比例本身應隨時間變動**。

::: {.callout-note}

### 什麼是 OU 過程

**Ornstein–Uhlenbeck 過程**是連續時間的均值回歸模型：

$$dX_t = 	heta\,(\mu - X_t)\,dt + \sigma\,dW_t$$

- $\mu$：長期均衡水準　- $	heta > 0$：**回歸速度**　- $\sigma$：波動幅度
- $dW_t$：標準布朗運動（隨機衝擊）

$	heta(\mu - X_t)$ 是**拉回項**：價差高於 $\mu$ 時為負、低於時為正，
且偏離越遠拉力越強。這正是配對交易的核心假設之數學形式——
**它把「價差終將回歸」寫成可估計、可檢定的模型**。

由 $	heta$ 可導出**半衰期** $HL = \ln 2 / 	heta$：價差消除一半偏離所需時間。
本研究於篩選層以 $1 \le HL \le 42$ 日作為濾網（§3.2），
即由此而來——回歸太慢的配對，在 126 日交易期內來不及收斂。

**優勢**：具明確的最適性定義。**代價**：真實動態偏離 OU 假設時，最適性不成立。

:::

**監督式學習**
**Krauss, Do & Huck (2017).** *Deep neural networks, gradient-boosted trees,
random forests: Statistical arbitrage on the S&P 500.*
**European Journal of Operational Research, 259(2), 689–702.**
以 DNN／GBT／隨機森林預測相對報酬——惟其架構已近於**橫斷面選股**，
而非傳統配對交易。
相關者另有 **Fischer & Krauss (2018)**, *Deep learning with long short-term
memory networks for financial market predictions*, **EJOR, 270(2), 654–669**。


## 深度強化學習路線

> **Kim, T., & Kim, H. Y. (2019).** *Optimizing the pairs-trading strategy using
> deep reinforcement learning with trading and stop-loss boundaries.*
> **Complexity, 2019, Article 3582516.**
>
> 以 **deep Q-network（DQN）** 同時最適化進場門檻與停損邊界，
> 動作空間為門檻組合的離散集合，報酬函數為交易損益，
> 報告優於固定門檻的傳統規則。

> **朱羿璁 (2025).**《結合共整合避險比率與混合獎勵設計之強化學習配對交易模型》
> 於台灣市場驗證類似架構。

**理論吸引力**：不需預測價格，僅需在給定狀態下**選擇門檻**——
問題結構與強化學習的「狀態—動作—報酬」框架天然契合
（**Sutton & Barto, 2018**, *Reinforcement Learning: An Introduction*, 2nd ed., MIT Press）。

::: {.callout-important}

### 本研究於此路線上的位置

沿用 Kim & Kim (2019) 的**動作空間設計**（自離散的門檻組合中擇一），
但**不**沿用其學習範式。

在本研究的設定下，每組配對每期只做一次決策、選定後交由固定規則執行整期，
故對歷史上任一配對期，把候選門檻各自模擬一遍即可得到**全部動作**的報酬。
學習問題因而是**全資訊監督回歸**（由狀態預測各動作報酬，取 $rg\max$），
**不存在探索與利用的取捨**，非部分回饋的強化學習（詳見 §3.4.4）。

本研究稱此交易端為 **DL-THR**，並另實作真正的部分回饋版本 **RL-THR**
（同一動作選單、同一狀態，僅改為只觀測所選動作的報酬 + $arepsilon$-greedy）
作為受控對照，用以估計反事實標籤本身的價值。

:::

::: {.callout-warning}

### 此一區分具文獻上的意義

既有研究在同類設定下沿用「深度強化學習」之名時，
**未必檢視其問題結構是否真的具有探索與利用的取捨**。

若動作報酬可完整回算，強化學習的機制優勢即不成立——
觀察到的增益應歸因於**門檻選擇這個決策形式**，而非強化學習演算法。

:::


## 交易成本與本研究的區隔（交易期）

**Do & Faff (2012)** 估計美股配對交易單邊成本約 **30 個基點**（含佣金與市場衝擊），
並主張報告 **break-even 成本**，以使結論不依賴單一成本假設。

::: {.callout-note}

### 什麼是 break-even 成本

**使策略淨利恰好歸零的那個交易成本水準。**

策略的毛利固定，成本則隨費率線性增加。把費率一路調高，淨利必然遞減並在某點
穿越零——**該點即 break-even 成本** $c^*$。

$$c^{*} = 	ext{現行單邊費率} \;+\; rac{	ext{淨利}}{\Sigma\,	ext{名目交易額}}$$

**為什麼要報它**：交易成本是**假設**而非觀測值，不同研究者的估計可差一倍以上。
只報「在 29bp 假設下淨利為正」，讀者無從判斷該結論有多脆弱；
報 $c^*$ 則直接回答「**成本要漲到多少，結論才會翻轉**」。

$c^*$ 與現行假設的距離即**成本餘裕**——餘裕越薄，結論對成本假設越敏感。

:::

### 本研究與既有文獻的區隔

**一、同一批配對上的單變因對照**——固定形成期產出，僅替換交易端。

**二、增益來源的分解**——既有文獻多僅報告學習式交易端是否較優；
本研究將 DL-THR 的動作還原後，分解為「**拒絕交易**」與「**調整門檻**」兩條管道，
並對各管道設置對照。

**三、成本敏感度的量化**——依 Do & Faff (2012) 建議報告 break-even 成本，
而非僅在單一假設下比較。


# 2.4 推論方法的文獻

本研究的主檢定並非既有配對交易文獻的慣例，故其依據須一併回顧。

## 慣例的問題：以參數網格為觀測單位

配對交易研究普遍在**參數網格**上報告績效（如 `top_n` × 停損的多種組合），
並以這些配置作為觀測單位進行跨策略檢定。

::: {.callout-important}

### 這構成偽重複（pseudo-replication）

概念出自生態學的實驗設計批評（**Hurlbert, 1984**, *Pseudoreplication and the
design of ecological field experiments*, **Ecological Monographs, 54(2), 187–211**）：
**當多個「觀測」其實共用同一個實驗單元時，把它們當成獨立樣本會虛增自由度**，
使 $p$ 值系統性偏小。

配對交易的情形完全對應——15 個參數配置共用**同一份資料、同一段期間、
同一批配對**（`top_n`=10 與 20 共用 10 組配對；三種停損只是同一批交易的
不同出場規則）。有效樣本數接近**一條回測路徑**，而非 15。

:::

**本研究因此改以「時間」為抽樣單位**：檢定的對象是逐日報酬差序列，
樣本數為交易日數（6,287），而非參數配置數。


## 主檢定：循環 block bootstrap

改以時間為抽樣單位後，新的問題是**自相關**：本研究滾動步長（21 日）小於
交易期長度（126 日），任一時點有 6 個交易期同時運行，逐日報酬差因而序列相關。
標準 bootstrap 假設觀測獨立，直接套用會低估標準誤。

> **Künsch, H. R. (1989).** *The jackknife and the bootstrap for general stationary
> observations.* **The Annals of Statistics, 17(3), 1217–1241.**
> 提出 **moving block bootstrap**：改抽**長度 $L$ 的連續區塊**而非單點，
> 區塊內部的相關結構被完整保留，只有區塊之間被打散。
>
> **Politis, D. N., & Romano, J. P. (1992).** *A circular block-resampling procedure
> for stationary data.* 將序列**首尾相接成環**後再抽區塊，
> 使每個觀測被抽中的機率相等，消除 Künsch 版本中端點觀測被低估的偏誤。

::: {.callout-note}

### 為何選它作為單一主檢定

| | Newey-West HAC | 循環 block bootstrap |
| :--- | :--- | :--- |
| 分布假設 | 漸近常態 | **無** |
| 研究者自由度 | **落後階**須自行選定並辯護 | 無（$L$ 由交易期長度直接給定） |
| 日報酬厚尾／偏態 | 近似可能失準 | **不受影響** |
| 輸出 | $p$ 值 | $p$ 值**與信賴區間**（同一次重抽） |

兩法在本研究全部 14 組對照上結論完全一致，故**換用的理由不在結果而在可辯護性**。
HAC 保留為各表的對照欄。

:::

**兩項配套程序**

> **Benjamini, Y., & Hochberg, Y. (1995).** *Controlling the false discovery rate.*
> **JRSS-B, 57(1), 289–300.** —— 同一命題涉及多組對照時控制偽發現率。
>
> **Bailey, D. H., & López de Prado, M. (2014).** *The deflated Sharpe ratio.*
> **Journal of Portfolio Management, 40(5), 94–107.** —— 校正「自 $N$ 次試驗中
> 挑最佳」的選擇偏誤。本研究以**全網格等權組合**為報告口徑使該偏誤自源頭消失，
> 故 DSR 僅列於附錄作為對照。


# 2.5 文獻缺口與貢獻定位

## 三項缺口

::: {.callout-important}

**其一：跨方法比較的推論基礎薄弱**
多數研究以各自參數網格的**最佳值**互相比較，未控制選擇偏誤；
且以參數組合為觀測單位的檢定，觀測間高度相關，**不滿足獨立性假設**。

**其二：形成期與交易期的改良常被混合施加**
當一項研究同時更換分群方法、篩選規則與交易端時，
其績效差異**無法歸因至任一單項**。

**其三：機器學習分群與共整合篩選的相容性未被檢驗**
兩者對「好配對」的定義不一致，而多數實作將兩者並用，**未討論其交互作用**。

:::

## 本研究的貢獻定位

| 缺口 | 本研究的填補方式 |
| :--- | :--- |
| 推論基礎 | 以**時間維度**的統計檢定取代參數網格檢定（逐日差分 + block bootstrap），並一律報全網格等權組合 |
| 歸因性 | 以**中性組裝器**實現單變因對照，每次僅動一個變因 |
| 交互作用 | 對「**分群 × 篩選**」設置受控消融，含「不分組」零點 |
