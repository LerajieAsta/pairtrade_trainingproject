# 2.1 配對交易

**獲利假設建立於相對定價**：不主張能預測個股價格，
只主張兩檔具共同經濟驅動因子的證券，其相對價格應維持在某均衡附近。

GGR (2006) 實證確認其配對組合對 Fama-French 三因子曝險極低
→ 報酬難以由既有風險因子解釋。

> **此機制並未要求兩檔證券同屬一個產業。**
> 產業一致性是共同經濟衝擊的**充分而非必要**條件——
> 此區別在機器學習分群文獻中成為核心爭點。

## GGR 的基準設計——與一個常被忽略的性質

1. 形成期 12 個月選對、交易期 6 個月
2. 正規化價格路徑（累積總報酬指數，首日 = 1）
3. 距離準則：SSD 最小的前 N 組
4. **開倉門檻：偏離 2 倍標準差進場、回歸至零平倉**
5. 對沖比例固定為 1

> **第 4 項其實定義了一個動作空間——一個只有單一元素的動作空間。**
>
> 學習式交易端的核心變化不在於「會學習」，而在於**把這個集合擴大**；
> 而集合要擴大成什麼形狀，是一個尚未被系統檢視的設計選擇。

## 報酬衰減：一致的實證發現

| 研究 | 發現 |
| :--- | :--- |
| Do & Faff (2010) | 獲利自 1990 年代末起顯著下降；**大型股衰減更明顯** |
| Do & Faff (2012) | 納入單邊約 30 bps 成本後，多數策略淨報酬**不再顯著** |
| Krauss (2017) | 五大類方法報告績效在 2000 年代後普遍下降；**方法間差異小於樣本期間造成的差異** |

**本研究的實證期間 2000–2025 完全落在衰減已被記錄之後**——
這是刻意的：問題不是「配對交易是否曾經獲利」，
而是「學習式交易端能否在衰減環境下重建優勢，以及其效果來自何處」。

# 2.2 配對選擇方法的演進

**在本研究中，此層不是被檢定的處理，而是提供五個配對來源。**

| 方法 | 相似性的定義 | 對價差性質的要求 |
| :--- | :--- | :--- |
| 距離法（GGR 2006） | 歷史價格路徑接近 | 無 |
| 共整合法（Vidyamurthy 2004） | 存在定態線性組合 | 定態（＋半衰期、Hurst） |
| Copula 法（Liew & Wu 2013） | 聯合分布的相依結構 | 依 Copula 設定 |
| **機器學習分群** | **特徵空間的鄰近性** | **通常不作要求** |

許鈞翔 (2025) 另以 **DTW**（Sakoe & Chiba 1978）改良距離度量，
容許時間軸非線性對齊。

## 非監督式機器學習分群：三篇代表

| 研究 | 母體 / 頻率 | 結果 |
| :--- | :--- | :--- |
| **Avellaneda & Lee (2010)** | 美股 | 以報酬相關矩陣主成分萃取共同因子，殘差建訊號 |
| **Sarmento & Horta (2020)** | 208 檔商品 ETF／**5 分鐘** | 分群 3.79 > 不分組 3.58 > 依類別 2.59 |
| **Han, He & Toh (2021)** | CRSP 全美股／月頻 | Agglomerative 最佳：年化超額 **24.8%**、Sharpe 2.69 |

Han et al. 的兩項關鍵發現：

- 獲利部分來自**跨產業**的價差發散
- 獲利**非由小型股驅動**（剔除規模後 20% 分位僅小幅降低）

## 一項未被充分討論的張力

> **機器學習分群與共整合篩選對「什麼是好配對」的定義並不一致。**
>
> 前者以特徵相似性為準，後者要求價差平穩性；
> 特徵相近的跨產業股票未必具備使價差回歸的共同衝擊。

- Han et al. (2021) 的設計中**並無**共整合篩選
- 多數以 GGR 為基礎的實作則**保留**之
- 兩者的組合是否相容，**文獻未見討論**

## 本研究對此張力的處理已改變

本研究早期版本將「分組方法」設為被檢定的處理變數，
以九組直接對照檢驗資料驅動分群是否優於 GICS。

**結果為方向一致的 null**（9/9 偏向 GICS、校正後 0/9 顯著），
而其**最小可偵測效果大於參照臂自身的年化報酬**
→ 該設計的檢定力**不足以支持任何方向的結論**。

> 故本研究不再以此為主張：五種分組方式視為**配對來源**，
> 其相對優劣列為描述性觀察（第五章 5.3），
> 上述相容性張力留作後續研究。

# 2.3 交易端的規則設計與動作空間

| 路線 | 代表 | 要點 |
| :--- | :--- | :--- |
| 固定門檻 | GGR (2006) | 門檻**外生給定且固定** |
| 隨機控制 | Elliott et al. (2005) | OU 過程 ＋ Kalman，推導最適邊界；需分布假設 |
| 監督式學習 | Krauss et al. (2017) | 預測相對報酬；已近於橫斷面選股 |
| **深度強化學習** | **Kim & Kim (2019)**、朱羿璁 (2025) | 動作空間為門檻組合的離散集合 |

固定門檻的限制：最適門檻應隨波動度、回歸速度與市場狀態變化，
而固定參數在低波動期過於保守、高波動期過於激進。

## 2.3.5 以動作空間重讀這一支文獻

上述各法通常依「用了什麼演算法」分類。
**本研究主張另一個分類軸更具解釋力：代理人被允許做什麼決策。**

| 動作空間 | 每個交易期的決策次數 | 代表 |
| :--- | :--- | :--- |
| 單一常數（不決策） | **0** | GGR (2006) 的 2.0／0.0 |
| 每期選一組門檻 | **1** | Kim & Kim (2019) |
| 逐日決定持倉 | **≈ 126**（交易期長度） | 一系列 DRL 實作 |

> 三者的差異不是演算法強弱，而是**決策的解析度**，且相差兩個數量級。
> 解析度高者在理論上**涵蓋**解析度低者
> （逐日決策可模擬任何固定門檻規則）。

## 但表達能力不是唯一的考量

> **取捨：動作空間的表達能力 vs 每次決策的訊噪比與成本**
>
> 決策次數愈多 → 每次決策所依據的訊號愈短、噪音佔比愈高，
> 且每次決策都可能產生一次**換手成本**。

此取捨在既有文獻中**未被當作實驗變因處理**，三項後果：

1. 失敗時，無法判斷來自演算法或動作空間
2. 成功時，無法判斷增益來自「逐對辨識」或「動作空間的形狀」
3. 各研究頻率與母體不同 → 跨研究比較無法替代同一管線內的消融

Do & Faff (2012) 建議報告 **break-even 成本**——
此建議與上述取捨直接相關：解析度提高的代價正由成本承擔。

## 本研究於此路線上的位置

> 本研究**沿用 Kim & Kim (2019) 的動作空間設計**（自離散門檻組合中擇一），
> 但**不沿用其學習範式**。
>
> 在本研究的設定下，歷史配對期中**全部動作的報酬皆可精確反事實回算**，
> 故該學習問題為**全資訊監督回歸**而非部分回饋的強化學習。
> 本研究稱此交易端為 **DL-THR**，並另實作真正部分回饋的 **RL-THR** 作對照。

**此區分的文獻意義**：既有研究在同類設定下沿用「深度強化學習」之名時，
未必檢視其問題結構是否真的具有探索與利用的取捨。
若動作報酬可完整回算，RL 的機制優勢即不成立。

# 2.4 一項跨文獻的觀察：AI 成功案例的頻率相依

| 研究 | 方法 | **資料頻率** | 報告績效 |
| :--- | :--- | :--- | :--- |
| SAPT (2021) | 小波 CNN+LSTM 預測結構性斷裂 ＋ DQN 停損 | **分鐘級** | 較基準 +456% |
| HDRL-Trader (2022) | 雙網路（動作 ＋ 停損邊界） | **日內** | 平均報酬 82.4% |
| Sarmento & Horta (2020) | OPTICS 分群 | **5 分鐘** | Sharpe 3.79 |
| Kim & Kim (2019) | DRL 門檻 ＋ 停損 | **日內** | 優於固定門檻 |
| Han et al. (2021) | 特徵空間分群（CRSP 全美股） | 月頻 | 年化超額 24.8% |

> **AI 於配對交易的成功案例，其資料頻率高度一致地集中在日內。**
> 唯一的日頻以上例外（Han et al.）依賴遠大的母體與遠豐富的特徵集。

## 兩種相反的解讀，與本研究能提供的東西

| 解讀 | 若為真，則 |
| :--- | :--- |
| AI 方法有效，**日頻訊噪比不足** | 日頻的失敗是對**資料條件**的陳述 |
| 報告績效與頻率無關，**型態出於發表選擇** | 日頻的失敗是對**方法**的實質否證 |

本研究無法在單一研究內分辨，但可提供該型態所缺的：**日頻上的受控對照**。

### 兩項同構想、異頻率的直接測試（皆未成立）

- **SAPT 的結構性斷裂構想**：日頻 S&P 500 上，模型對 `diverged` 標籤的
  AUC **0.5286** < **SSD 排序本身 0.5393**
- **Kim & Kim 與 SAPT 的 σ 停損**：四條主力臂 × 三個水準，
  **無一使年化報酬為正**；`NOGRP-SDP` 自 **+1.452%** 降至 −0.627／−0.873／−0.090%，
  Profit Factor **1.191 → 0.83~0.93**（跌破 1）、利用率 **66.2% → 12.7%**

# 2.5 文獻缺口與本研究的貢獻定位

| # | 缺口 |
| :--- | :--- |
| 一 | **動作空間未被當作實驗變因**——各自選定並報告績效，無人在同一管線內只變動它 |
| 二 | **增益來源未被分解**——未區分「逐對辨識能力」與「動作空間的形狀」 |
| 三 | **成功案例的頻率相依未被指出**——集中於日內，兩種解讀未被分辨 |

**本研究的填補**：

1. 以**預先登記**的受控消融，把動作空間設為唯一變因
2. 以「逐對學門檻 vs 擴張窗最佳常數」分離辨識能力與動作空間形狀
3. 提供日頻上的受控對照，並直接測試兩項來自日內研究的核心構想
