# 本章要處理的三個遺留問題

第四章的三段證據回答了三個研究問題，但留下三個必須另行處理的問題：

1. 交易層的相對效果通過了多重檢定校正——**那它為什麼仍不可交易？**（5.1）
2. 若配對交易的多數交易確實收斂獲利，**獲利被什麼吃掉了？**（5.2）
3. 五個配對來源之間有明顯的績效差異，**這件事能說到什麼程度？**（5.3）

以研究限制作結（5.4）。

# 5.1 為何統計顯著仍不可交易

4.3 節的結論是「DL-THR 相對固定門檻，五個來源方向全正、BH 校正後 3/5 顯著」，
這是本研究**唯一通過多重檢定校正的正向結果**。

> **但它不足以支持任何可交易的宣稱。**
> 理由有三道，**彼此獨立，任一道都足以否定該宣稱**。

| # | 界線 |
| :-- | :--- |
| 一 | 絕對績效極低，且未達統計顯著 |
| 二 | Deflated Sharpe Ratio 無一通過 |
| 三 | 成本餘裕的離散大於其中心 |

## 5.1.1 第一道：絕對績效極低，且未達統計顯著

主軸 15 格的等權年化（**帳面口徑**，15 個參數配置等權組合）：

| 分組 | SSD | DTW | SSD-DTW-PCA |
| :--- | ---: | ---: | ---: |
| 不分組 | +0.641% | **+0.780%** | +0.607% |
| GICS 產業 | +0.345% | +0.499% | +0.541% |
| Agglomerative | +0.040% | +0.358% | +0.310% |
| HDBSCAN | −0.287% | +0.269% | +0.076% |
| K-means | −0.484% | −0.330% | −0.343% |

疊加 DL-THR 後最佳為 GICS-SSD：帳面年化 **+1.026%**、25 年終值 13,062、Sharpe 0.276。

## 「年化」有三個口徑——先前版本混用了兩個

靜態槽位的分母是理論上限，該臂平均資金利用率僅 **42.6%**（逾半數資金長期閒置）：

| 口徑 | GICS-SSD＋DL-THR | GICS-SDP＋DL-THR |
| :--- | ---: | ---: |
| 帳面年化（閒置現金計 0%） | +1.026% | +1.016% |
| 動用資本年化 | +2.793% | +2.697% |
| **rf 超額年化**（閒置現金計息後超過 rf 的部分） | **+0.375%** | **+0.390%** |
| 平均資金利用率 | 42.6% | 43.2% |

> **一項已更正的不一致。** 先前的陳述是「僅持有無風險資產 10,000 將成為 16,400，
> 本研究的任何配置皆未達此水準」——該比較把**策略的閒置現金當 0% 計息，
> 卻讓對照組整筆吃 2%**。
>
> 一致口徑重算：組合總報酬約 **2.375%**，即**高於**無風險利率 **0.375 個百分點**。
> `Excess_Ret_RF` 自始即在資料庫中，但先前未進入正文。

## 以 rf 超額重看那 15 格——結論比帳面口徑更強

| 分組 | SSD | DTW | SSD-DTW-PCA |
| :--- | ---: | ---: | ---: |
| 不分組 | −0.308% | **+0.009%** | −0.391% |
| GICS 產業 | −0.705% | −0.522% | −0.471% |
| Agglomerative | −1.042% | −0.689% | −0.771% |
| HDBSCAN | −1.380% | −0.822% | −1.015% |
| K-means | −1.577% | −1.435% | −1.456% |

（利用率一律落在 56.2%–57.6%）

> **15 格固定門檻基準，無一在 rf 超額口徑下為正。**
> 最高者不分組 DTW 僅 **+0.009 pp**，實質為零。
>
> 亦即**單靠配對選擇與固定門檻，這 25 年連無風險利率都跑不贏**；
> 唯二為正的是疊加 DL-THR 的兩條 GICS 臂（+0.375%、+0.390%）。

## 但 DL-THR 的優勢有一部分是機械的

DL-THR 會 SKIP 三至四成配對期，利用率只有 42.6%／43.2%（固定門檻約 57%）
→ 被扣掉的 `rf × 利用率` 較小，rf 超額被墊高。

| 配對來源 | Δ rf 超額 | 真實報酬差 | 閒置現金多收的 rf |
| :--- | ---: | ---: | ---: |
| GICS-SSD（DRL − Z-Score） | +1.080 pp | **+0.798 pp** | +0.282 pp |
| GICS-SDP（DRL − Z-Score） | +0.861 pp | **+0.579 pp** | +0.282 pp |

> **拆出來的「真實報酬差」與 4.3 節的年化Δ逐位相同**（+0.798、+0.579）
> ——一項內部一致性檢查。
>
> **那 +0.282 pp 不應計為技巧**：它只是「少投入資本、多收利息」，
> 與 4.4 節「總曝險減少」那項替代解釋是同一件事的另一面。

## 此更正使第一道界線變弱，但不使其消失

**其一，rf 超額口徑在機制上獎勵低利用率。**
它等於「算術年化 − rf × 利用率」，故只動用兩成資本的策略，
其餘八成的 rf 利息會直接計入。實測即有此例：

> `NOGRP-DTW-VG67` 的 rf 超額 **+0.499%** 高於兩條 DL-THR 臂，
> 但其資金利用率僅 **20.4%**。
>
> **該數字描述的是「組合的總報酬」，不是「配對交易這個動作本身賺了多少」。**

**其二，+0.375 個百分點的超額，其 Sharpe 只有 0.276**，低於 DSR 門檻 0.408。

此更正改變的是「有沒有跑贏無風險」這個二分判斷，
**不改變「報酬與其承擔的變異不相稱」這件事**。

## 且絕對績效未達統計顯著

對六組（三個 ML 分群來源 × 兩個交易端）檢定「日均報酬是否為零」：

> **6/6 不顯著**（$p$ = 0.186–0.850）

**4.3 節通過校正的是「相對於 Z-Score 的差」，不是這裡的絕對水準。**

兩者不可混為一談——這是本研究最常被誤讀的一處，故在此重述。

## 5.1.2 第二道：DSR 無一通過

若改以「網格最佳配置」報告，須扣除選擇偏誤（Bailey & López de Prado, 2014）。
試驗宇宙 53 個策略（config 口徑 1,392 格），門檻 $SR_0$ = 0.408：

| 策略 | 最佳配置 | Sharpe | DSR | 通過 0.95 |
| :--- | :--- | ---: | ---: | :---: |
| 不分組 × DTW | Top1/SL0% | 0.529 | **0.738** | ✘ |
| 不分組 × SSD | Top3/SL0% | 0.455 | 0.598 | ✘ |
| 不分組 × SDP | Top3/SL0% | 0.436 | 0.556 | ✘ |
| GICS × SDP | Top3/SL0% | 0.366 | 0.412 | ✘ |
| GICS × SSD | Top5/SL0% | 0.355 | 0.392 | ✘ |

**41 條策略無一通過 0.95。** 命題二的六個主角同樣全數未過
（Sharpe 0.049–0.281，DSR 0.036–0.260）。

> **即 4.3 節通過多重檢定校正的那三組，在選擇偏誤校正下仍不足以宣稱技巧。**

## DSR 最高者正好是一個反例

形成窗 504 日的對照臂 `GICS-SSD-FW504`（Top3/SL0%）：
全期 Sharpe **0.634**、DSR **0.885**，為全部 53 個方法之冠，仍未達 0.95。

> 之所以不列為主結果，是因為它**後半期 Sharpe 為 −0.389**（前半 +0.998）
> ——該值之高全部來自前半期，**以全期 Sharpe 挑選即會選中它**。
>
> **DSR 不足以獨力防止選擇偏誤**：它校正的是「看過幾個候選」，
> 不校正「以全期指標挑選已在後半期失效的策略」。

## 前後半分割的全樣本證據

把 6,287 個交易日自中點對半切開，對 491 個可分割的基準格各算兩次年化 Sharpe：

| 觀察 | 數值 |
| :--- | ---: |
| **兩半皆為正 Sharpe 的格數** | **29 / 491（5.9%）** |
| 以**全期** Sharpe 取前 10：後半為負者 | 8 / 10（後半中位 −0.074） |
| 以**前半** Sharpe 取前 10：後半為負者 | **10 / 10**（後半中位 −0.608） |

> **以任何全樣本指標挑選，挑到的幾乎都是後半期已失效的策略。**
> 全期前 12 名中僅兩格兩半皆正。

## 試驗宇宙的口徑是一項判斷，須明說

本研究採 $N$ = 53（method 口徑），為管線修正後重建之結果集的**實地清點**。

- 修正前曾跑過的 1,939 個配置已隨舊資料庫刪除、未在本文報告任何數字，故不計入
- 更保守的讀法會一併計入——該口徑下 **DSR 只會更低，本節結論不變**

> **動作空間消融的五條臂不計入試驗宇宙。**
> 它們不是為了挑選最佳策略而試的配置——三條為被否證的對照組，
> 且預先登記寫定不參與任何選擇（附錄 D）。
> **此為口徑宣告，不是沉默的省略。**

## 5.1.3 第三道：成本餘裕的離散大於其中心

往返 break-even 成本與現行假設 0.58% 的餘裕：
全距 **−15.2 ~ +43.0 個基點**，橫跨 58 bps，**恰與成本假設本身同量級**。

- K-means 來源兩臂餘裕皆為負 → 在現行成本假設下不具可行性
- 餘裕最高者為 **GICS ＋ DL-THR**（+41.3、+43.0 bps），
  break-even 達 0.99–1.01%，對 0.58% 有**逾七成餘裕**

> **故成本不是全篇一致的否定理由**，嚴格說只適用於 K-means 那條臂。
> 本研究不建議實務應用的理由主要在 5.1.1 與 5.1.2，**不在成本**。

## 5.1.4 三道界線的共同意涵

| 界線 | 內容 | 是否全面成立 |
| :--- | :--- | :--- |
| 絕對績效 | 帳面年化 1.026%；rf 超額僅 +0.375 pp 且 Sharpe 0.276；6/6 不顯著 | **全面** |
| DSR | 41 條無一通過 0.95；最高者後半期為負 | **全面** |
| 成本餘裕 | 全距 58 bps；K-means 為負，GICS＋DL-THR 逾七成餘裕 | 僅部分 |

> **「相對於某個基準較優」與「本身可交易」是兩個不同的主張，
> 而本研究只支持前者。**

這不是修辭上的保留，而是三道獨立證據的結果：
效果存在、方向一致（5/5）、機制可歸因（三來源排除三項替代解釋），
但水準太低（rf 超額僅 +0.375 pp，Sharpe 0.276）、選擇偏誤校正後不足、
且其中一個來源連成本都付不起。

> **完全漏掉 `STOP_LOSS_TRIGGERED`**，

收斂獲利 +16,217、期末強平 −9,369、停損 −5,776
→ 淨額僅 +1,072，**不到收斂獲利的十五分之一**。

## 5.2.1 強平的配對不是「尚未回歸」

若期末仍持倉者只是需要更多時間，延長交易期即可解決。
取主軸 15 臂的 `Top1/SL0%` 格（無停損，強平事件不被截斷），
對 3,107 筆期末強制平倉往後追蹤 126 個交易日：

| 追蹤期 | 21 日 | 42 日 | 63 日 | **126 日** |
| :--- | ---: | ---: | ---: | ---: |
| 累計回歸比例 | 17.5% | 25.7% | 33.2% | **45.4%** |

平倉當時 $|z|$ 中位為 3.51。再給一個完整交易期，**僅 45.4% 回歸**；
未回歸的 54.6%，其 $|z|$ 自 4.57 繼續擴大至 **6.82**。

> **這些配對不是尚未回歸，而是持續發散。**
> 即使把「回歸」放寬到 $z$ 曾觸及 0 即算，仍有超過一半永不回頭。
> 三項針對此現象的交易端改良皆被否證（附錄 C）。

## 5.2.2 強平率與績效——一個必須避開的陷阱

| 層級 | 相關係數 |
| :--- | :--- |
| 逐臂（15 臂） | Pearson $r$ = **−0.819**（$p$ = 0.0002）；Spearman $\rho$ = −0.896 |
| 逐配置內、跨 15 臂 | **15 個配置中 15 個為負**，中位 $r$ = −0.747 |

> **不可以全樣本混合計算此相關。** 225 個配置混合後 $r$ = **+0.483**，符號相反。
>
> 此為停損維度造成的 **Simpson 悖論**：停損 5% 的配置強平率僅 23.9%
> 但平均 Sharpe −0.233；停損 0% 強平率 45.0% 而 Sharpe +0.183。
> **必須逐配置比較，或以配置為區塊。**

**三項限制併陳**：(a) 15 臂不獨立 → $p$ 值偏樂觀；
(b) 相關非因果；(c) 描述性證據，非機制的正式檢定。

# 5.3 配對來源的影響（描述性）

**這件事能說到什麼程度——答案是：只能說到描述，不能說到檢定。**

| 分組方式 | 排除標的 | 期末強平率 | 平均 Sharpe | 等權年化 |
| :--- | ---: | ---: | ---: | ---: |
| **不分組** | **0%** | **33.5%** | **+0.109** | **+0.676%** |
| GICS 產業 | 15.1% | 36.1% | +0.062 | +0.461% |
| HDBSCAN | 26.4% | 36.7% | −0.053 | +0.019% |
| Agglomerative | 35.3% | 37.9% | +0.010 | +0.236% |
| K-means | 42.0% | 39.0% | −0.179 | −0.386% |

趨勢與期末強平率一致（33.5% → 39.0%），
但**並非嚴格單調**：Agglomerative 排除較多卻表現較好。**故為趨勢而非定律。**

## 逐格觀之，以及一條不穩固的分離

| 分組方式 | SSD | DTW | SSD-DTW-PCA |
| :--- | ---: | ---: | ---: |
| **不分組** | **+0.110** | **+0.105** | **+0.112** |
| GICS 產業 | +0.043 | +0.068 | +0.076 |
| Agglomerative | −0.040 | +0.028 | +0.042 |
| HDBSCAN | −0.129 | +0.012 | −0.044 |
| K-means | −0.215 | −0.158 | −0.165 |

不分組三格全部優於同排序的 GICS，GICS 三格又全部優於任一資料驅動分群格——

> 惟 GICS 的最低格（+0.043）僅高出資料驅動分群的最高格（+0.042）
> **0.001 個 Sharpe 單位**。該分離已無實質餘裕，**不應視為穩健**。

## 為何這一節只能是描述性的

早期版本曾以九組直接對照檢定「資料驅動分群是否優於 GICS」：
方向一致的 null（9/9 偏向 GICS、校正後 0/9 顯著）。

> 該檢定的**最小可偵測效果為 1.14 個百分點，
> 大於參照臂自身的年化績效（+0.461%）**——
> 對 0.3 個百分點的真實效果，**檢定力僅 11%**。

**成因是結構性的**：形成期層的兩臂必然持有**不同的配對集合**，
差分中混入兩套標的的特異變異；交易期層兩臂共用同一批配對，差分即消噪。
標準誤相差 1.83 倍 → 形成期層需 3.3 倍樣本；
若要求 MDE 降至 0.3 pp，約需 **14.5 倍樣本（逾三個半世紀的日資料）**。

> **因此本研究對配對來源之間的優劣不作統計宣稱。**
> 唯一的出路是在形成期層構造**配對設計**（兩臂共用同一批標的而僅變動分組）
> ——本研究未能設計出這樣的對照，列為後續研究的首要方法論課題。

## 一項附帶觀察：篩選層的價值取決於候選池規模

- **不分組**時施加共整合篩選：等權年化 +0.306% → **+0.641%**
- **Agglomerative 分群且移除產業先驗**後施加篩選：+0.180% → **−0.140%**

> **候選池已被切小時，篩選進一步壓縮可選範圍，其代價超過其收益。**

此即第二章 2.2.5 節所指的**相容性張力**，同樣為描述性觀察
（對應的 14 組正式對照經 BH 校正後無一顯著，校正後最小 $p$ = 0.363）。

# 5.4 研究限制（依對結論的影響程度排列）

| # | 限制 |
| :-- | :--- |
| 一 | **動作空間消融的範圍**：單一來源／格點／口徑、108 期、未固定種子、各跑一次 |
| 二 | **母體範圍**：S&P 500 而非 CRSP 全美股，方向明確但無法實驗 |
| 三 | **特徵維度**：自 XBRL 建出 40 個僅 12 個過門檻；2000–2008 無財報特徵 |
| 四 | **存活者偏誤未完全消除**：170 檔已下市中 49 檔（29%）無價格資料 |

其一的量級可對照：效果量（換手 2.1–4.1 倍）**大於**已量測的執行間雜訊
（同組態兩次執行全距 2.0–2.8 倍），兩條基準臂則逐位確定；
但**無法排除「該結論在其他來源或格點上量級不同」**。

其四的偏誤方向明確：樣本偏向倖存者、絕對績效被高估
→ 故 5.1.1 的「絕對績效不顯著」**偏保守**。
