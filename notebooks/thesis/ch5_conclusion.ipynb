# 5.1 研究發現綜整

## 先看要部署的那個系統

本研究將配對交易的改良拆解為**形成期的分組方法**與**交易期的門檻選擇**兩層，
分別檢驗其貢獻，並在**系統層次**檢驗兩者的組合。先報告系統層次的結果。

> **組合系統**　252 日窗、21 日滾動的動態分群 ＋ **DL-THR 門檻選擇**交易端
> **傳統基準**　GICS 產業分組 ＋ **固定門檻 Z-Score**

**六組對照中五組於 BH-FDR 校正後達 5% 顯著**，效果量 **+0.63 ~ +1.51 pp**，
五組的信賴區間下界皆嚴格大於零。未達顯著者為全期的 K-means（BH $p$ = 0.186）。

::: {.callout-warning}

### 這是**相對**主張，必須與下一句同時陳述

同一批系統的**絕對**績效在等權口徑下**六組全部不顯著**，
信賴區間寬達 **2.4 ~ 2.9 pp**，年化 Sharpe 僅 **0.09 ~ 0.21**。

兩者虛無假設不同（前者「兩交易端績效相同」，後者「平均日報酬為零」），
故不矛盾——本研究能主張的是**方法之間的相對優劣**，
**不是**該策略可實現超額報酬。

:::


## 成分分解：功勞歸給哪一半？

| 期間 | 分群法 | 總效果 | BH 校正 $p$ | DL-THR 成分 | 分群成分 |
| :--- | :--- | ---: | ---: | ---: | ---: |
| 全期 | Agglomerative | **+1.156** | **0.0141** ✔ | **+0.787** ✔ | +0.369 ✘ |
| 全期 | HDBSCAN | **+1.025** | **0.0141** ✔ | **+0.865** ✔ | +0.160 ✘ |
| 全期 | K-means | +0.631 | 0.1864 ✘ | **+0.606** ✔ | +0.025 ✘ |
| 2012+ | Agglomerative | **+1.511** | **0.0074** ✔ | **+0.743** ✔ | +0.768 ✘ |
| 2012+ | HDBSCAN | **+1.194** | **0.0074** ✔ | **+1.000** ✔ | +0.195 ✘ |
| 2012+ | K-means | **+1.193** | **0.0203** ✔ | **+0.464** ✔ | +0.730 ✘ |

- **DL-THR 成分六組全部顯著**（+0.464 ~ +1.000 pp，$p$ = 0.0000 ~ 0.0200）
- **分群成分六組全部不顯著**（+0.025 ~ +0.768 pp，$p$ = 0.122 ~ 0.954）

::: {.callout-important}

### 顯著的是「系統」，不是「分群」

組合系統跨過門檻，是兩個方向相同的效果**相加**的結果。
分群成分方向一致為正（2012+ 的 Agglomerative +0.768、K-means +0.730，
數值上已接近同列的 DL-THR 成分），但單獨檢定無一顯著。

→ 可主張：**本研究所提的完整系統顯著優於傳統作法。**
→ 不可主張：**ML 分群有效。** 那需要分群成分自身顯著，而本研究做不到。

:::

> **正確的一句話**：完整系統顯著優於傳統基準，**但把功勞歸給哪一半，資料還答不出來。**
>
> **正確的一句話**：完整系統顯著優於傳統基準，**但把功勞歸給哪一半，資料還答不出來。**


## 分層歸因：兩個命題，方向相反的答覆

| | 命題 1（形成期） | 命題 2（交易期） |
| :--- | :--- | :--- |
| **主張** | ML 分群能建立更優的配對搜尋空間 | 門檻選擇優於固定門檻規則 |
| **結果** | **未獲支持** | **獲得支持** |
| **證據** | 9 組 BH 後無一顯著（最小 $p$=0.455），方向 ML 優 5 組／GICS 優 4 組 | 五配對底全部顯著（$p$=0.0000–0.0060） |
| **信賴區間** | 9 組**全部涵蓋 0**，寬 1.2–1.9 pp | 五組**全部落在零的右側**，最保守下界 +0.20 pp |
| **穩健性** | HAC 對照一致 | HAC 對照一致 |

::: {.callout-important}

### 命題 1 是「未能拒絕虛無假設」，**非**「GICS 顯著較優」

而且這個 null 的**鑑別力可以量化**，不必停在定性陳述：

| | $SE$ 中位 (pp) | MDE 中位 (pp) | 對 0.3 pp 真實效果的檢定力 |
| :--- | ---: | ---: | ---: |
| 命題 1（兩臂持**不同**配對） | 0.406 | **1.14** | **11%** |
| 命題 2（兩臂**共用**配對） | 0.214 | 0.60 | — |

若 ML 分群的真實優勢為 0.3 pp——已足以翻轉參照臂 −0.11% ~ −0.40% 的損益方向
——本檢定偵測到它的機率僅 **11%**。即使命題 1 為真，本實驗仍有近九成機率
報告「不顯著」。**故此一 null 對命題 1 的真偽幾乎不具鑑別力。**

:::

::: {.callout-note}

### 命題 2 的顯著則不是檢定力僥倖

點估計中位 +0.787 pp 明顯大於該比較自身的 MDE（0.60 pp）。
增益最大者為**傳統 GICS 配對底**（+1.105 pp）→ 此改良與「配對如何找到」正交。

:::


## 命題 2 的增益機制已釐清

**四項替代解釋皆被排除**

| 替代解釋 | 檢驗方式 | 結果 |
| :--- | :--- | :--- |
| 拉高進場門檻 | 同門檻（2.2）對照 | 五支仍顯著；門檻管道複製 **−8.9% ~ 28.4%**，五組中一組顯著 |
| 篩掉劣質配對 | SKIP 置換檢定 | **75 格僅 6 格顯著**（隨機期望 3.75） |
| 減少曝險 | 進場次數比較 | DL-THR 反為 Z-Score 的 **1.55–1.92 倍** |
| 全資訊設定 | RL-THR 部分回饋對照 | 兩臂**相當**（+0.015 pp，CI [−0.38, +0.45]，$p$=0.94） |

> ⚠ 門檻管道並非毫無貢獻：HDBSCAN 底的 ZS(2.2)−ZS(2.0) 達 +0.246 pp（$p$=0.014），
> 複製 28.4%。但 DL-THR 在同門檻下仍顯著勝出（+0.619 pp，$p$=0.0052），
> 故「增益不僅是拉高門檻」的結論維持（§4.2）。

殘差與「**槽位週轉**」假說一致 —— 惟需重跑回測方能直接驗證。

## 命題 1 的失敗不可歸因於實作選擇

三項候選原因逐一受控檢驗，**皆使形成期結構如預期改變，但績效皆未改善**：

1. **形成期設計** —— 產業 one-hot、共整合篩選、缺「不分組」零點
2. **交易機制** —— Han et al. 四項差異逐步施加
3. **特徵維度** —— 連續特徵 7 → 19 維


## 核心發現：形成期的改良難以驗證，且難在**設計**

::: {.callout-important}

### 兩層比較的不對等來自設計，而非資料量

命題 2 的兩臂**共用**同一批配對，逐日差分消去共同的市場衝擊；
命題 1 的兩臂持有**不同**的配對集合，差分中混入兩套標的的特異變異。

標準誤相差 **1.89 倍** → **等效於形成期層需 3.6 倍的樣本期間**。
若要求 MDE 降至 0.3 pp（能看見經濟上實質的差異），
形成期層約需 **14 倍**樣本、**逾三個世紀**的日資料。

更換分群演算法、增加特徵維度、改良插補方式**皆不影響它**。

:::

**輔證**：跨產業配對比例自 **0%（GICS）變動至 69%（完全不分組）**的六種設定中，
2012+ 績效全部落在 −1.84% ~ −0.66%，未觀察到可偵測的差異。
惟其全距（1.18 pp）**尚不及單一對照的區間寬度（1.18–1.91 pp）**
→ 此為**輔證，不是等價證據**（§4.5.2）。

**絕對獲利能力未獲證實**

以可事前實施的**全網格等權組合**計，六組配置年化落在 **−0.36% ~ +0.77%**，
bootstrap 檢定**無一顯著**，信賴區間寬達 **2.4 ~ 2.9 pp**。

> 此非樣本不足而是**結構性限制**：等權組合年化 Sharpe 僅 **0.09 ~ 0.21**，
> 在此量級下欲達統計顯著約需 **88 年以上**樣本。


# 5.2 研究貢獻

::: {.callout-warning}

### 先界定本節的口氣

以下三項**皆非新方法**。偽重複、單變因設計、信賴區間的解讀，
在統計與實驗設計文獻中都是既有的標準作法。

本研究的貢獻在於**把它們搬進配對交易的比較研究裡並貫徹到底**——
該領域的慣例與這些標準之間存在明確落差，而本研究記錄了落差的後果。

:::

## 方法論貢獻（一）：把偽重複問題帶進配對交易的跨策略檢定

配對交易文獻常以「參數網格」為觀測單位進行跨策略檢定。

> **偽重複（pseudo-replication）** 一詞出自 **Hurlbert (1984)**,
> *Pseudoreplication and the design of ecological field experiments*,
> **Ecological Monographs, 54(2), 187–211** —— 指**當多個「觀測」其實共用同一個
> 實驗單元時，把它們當成獨立樣本會虛增自由度**。該概念在生態學已是標準批評，
> 但**未見於配對交易文獻的檢定設計討論**。

本研究的 15 個參數配置正是此情形：共用同一份資料、同一段期間、同一批配對
（`top_n`=10 與 20 共用 10 組配對；三種停損是同一批交易的不同出場規則）
→ 有效樣本數接近**一條回測路徑**，而非 15。

| | 以參數網格為單位 | 以時間為單位 |
| :--- | ---: | ---: |
| 命題 2 效果量 | +0.732 pp | **+0.787 pp** |
| 推論正當性 | ✘ 獨立性假設不成立 | ✔ block bootstrap（$p$ 值 + 95% CI） |

**效果量幾乎不變，但推論的正當性完全不同。**

> **可主張者**：本研究記錄了「以參數網格為觀測單位」在配對交易脈絡下的具體後果，
> 並提供一個可照做的替代口徑。
> **不可主張者**：本研究「發現」了偽重複問題。


## 方法論貢獻（一）續：以區間寬度區分兩種「不顯著」

雙尾檢定不顯著時，文獻常逕行宣稱兩方法「表現相當」。

> 此一謬誤有明確的方法學來源可引：
> **Altman & Bland (1995)**, *Absence of evidence is not evidence of absence*,
> **BMJ, 311(7003), 485**；
> **Greenland et al. (2016)**, *Statistical tests, P values, confidence intervals,
> and power: a guide to misinterpretations*,
> **European Journal of Epidemiology, 31(4), 337–350**；
> **Wasserstein & Lazar (2016)**, *The ASA statement on p-values*,
> **The American Statistician, 70(2), 129–133**。

本研究據此主張：該判斷應交由**信賴區間的寬度**承擔。

| 情形 | 可下的結論 |
| :--- | :--- |
| 區間涵蓋 0，**兩端皆小**（相對於參照臂自身的績效量級） | 可討論「**相當**」 |
| 區間涵蓋 0，**兩端皆大** | 只能說**檢定力不足** |

::: {.callout-note}

### 與等價檢定的關係

正式作法是**等價檢定**（equivalence testing，如 **Lakens, 2017**,
*Equivalence tests*, **Social Psychological and Personality Science, 8(4), 355–362**），
但它要求事前指定容忍邊界 $\delta$，而配對交易並無公認的 $\delta$。

本研究因此**不宣稱做了等價檢定**，僅以區間寬度作定性區分——
這是較弱、但不需要辯護 $\delta$ 的作法。
故第四章對「配對選擇不影響報酬」一類主張一律不予採納（§4.5.2）。

:::


## 方法論貢獻（二）：把單變因原則實作為架構保證

**單變因對照本身是實驗設計的基本原則**（一次只變動一個因子，
可上溯至 **Fisher, 1935**, *The Design of Experiments*）。
本研究的貢獻不在原則，而在**把它做成程式架構的性質**。

將形成期拆解為四個可獨立替換的層：

```
特徵萃取  →  分組  →  群內排序  →  統計篩選
```

各層以標準化介面銜接，任兩策略之組態差異可被機械地檢查。

::: {.callout-important}

### 為何這件事值得記述

配對交易文獻中，「更換分群方法」的實作通常伴隨其他環節一併改動
（不同的排序準則、不同的篩選門檻），使績效差異**無法歸因至任一單項**。
本研究第二章已將此列為缺口之一。

把單變因做成**架構保證**而非人為約定後，本研究得以在命題未獲支持時
**系統性地逐一排除實作原因**——形成期設計、交易機制、特徵維度三項候選
全部受控檢驗且皆非原因（§4.4）——而不只是宣稱「方法無效」。

:::

> **可主張者**：一個使單變因對照可被機械驗證的實作架構。
> **不可主張者**：單變因設計是本研究提出的。

::: {.callout-warning}

### 這項保證也有一處例外，已於第三章揭露

排序層的替換同時改變了排序與篩選的**施行次序**，
故排序層之間的比較嚴格說並非單一變因（§3.2）。
本研究的兩個主命題落在分組層與交易端，皆固定排序後端，不受此影響。

:::


## 方法論貢獻（三）：限制的量化揭露與偏誤方向

配對交易文獻的限制章節多為定性陳述（「樣本有限」「成本假設可能低估」），
讀者無從判斷某項限制會使結論**偏哪一邊**、偏多少。

> 此一要求同樣有來源：**Harvey (2017)**, *Presidential address: The scientific
> outlook in financial economics*, **Journal of Finance, 72(4), 1399–1440**
> 對財務實證研究的可信度提出系統性批評，主張研究者應揭露完整的試驗歷程
> 與偏誤方向，而非僅呈現最終結果。

本研究對每項限制報告**具體數字**與**偏誤方向**：

| 限制 | 量化 | 偏誤方向 |
| :--- | :--- | :--- |
| 存活者偏誤 | 170 檔已下市股中 49 檔（29%）無價格，且缺失與下市狀態相關（43% vs 17%） | 絕對績效**被高估** → 「不顯著」結論**偏保守** |
| 成本假設年代 | 0.29%/邊取自 Do & Faff (2012) 對 1962–2009 的估計，套用於 2000–2025 | 摩擦成本長期下降 → **偏保守**；另報 break-even 使結論不依賴此假設 |
| 特徵覆蓋率 | 自 SEC XBRL 建出 40 個，僅 12 個通過 70% 門檻；另 5 個評價類卡在 57–70%、11 個永遠無法取得 | 命題 1 的特徵維度結論僅在「SEC 申報所能供給的範圍內」成立 |

::: {.callout-important}

### 方向已知的限制，不對稱地支持結論

三項主要限制**全部**指向「絕對績效被高估」。
本研究的絕對績效結論是「無一顯著」——限制若被修正，
該結論只會更強而不會翻轉。

反之，命題 1 的「未獲支持」則受母體範圍限制影響，**方向不明**，
故明確限定於本設定。

:::

> 本研究另將試驗宇宙實地清點並釘死（$N$ = 110，§4.6.3），
> 使 Deflated Sharpe（**Bailey & López de Prado, 2014**）的關鍵參數可被檢查，
> 而非由研究者事後選定。


## 實證貢獻

**其一：為機器學習分群的適用邊界提供一份反面觀察**

Han, He & Toh (2021，SSRN 工作論文) 以 CRSP 全美股、126 維特徵報告年化 **24.8%**
（扣 20bp 成本後 15.9%）。本研究在 S&P 500、SEC 可得特徵子集下未能複製，
且以受控消融排除三項實作原因。

> **此結果不否定原研究**，也**不是**對它的重製——母體與特徵條件差距過大。
> 可陳述者僅為：該方法的優勢**可能**依賴母體廣度與特徵維度，
> 在受限資料條件下**本研究未能觀察到**。
>
> 且本研究對命題 1 的檢定力不足（MDE 1.14 pp），
> 故這份反面觀察的證據強度**低於**一次成功的重製所能提供者。

**其二：釐清門檻選擇增益的來源與非來源**

排除三項行為面的替代解釋（非拉高門檻、非篩掉劣質配對、非減少曝險）
→ 「DL-THR 確有貢獻」建立在更窄、也更可信的基礎上。
特別指出：損益分解中的「SKIP 貢獻 31–52%」是**會計恆等式而非技巧歸因**。

再排除一項問題性質上的解釋：本研究另建 **RL-THR**——同一動作選單、同一狀態、
同一網路，僅把訓練標籤縮成「實際選中的那一個」並改採 $\varepsilon$-greedy——
使其成為真正的部分回饋問題。兩臂在等權口徑下**無法區分**
（+0.015 pp，95% CI [−0.38, +0.45]，$p$=0.94），且區間兩端皆遠小於總增益 +0.787 pp，
故為「相當」而非檢定力不足。

> **增益不來自「這個問題碰巧是全資訊的」。** 配合已證偽的逐日定位動作空間
> （三代真強化學習，中位 Sharpe −1.1 ~ −2.3），失敗與成功兩個方向指向同一結論：
> **關鍵是動作空間設計，不是訓練方法。**

此對照同時提供報告口徑的實例：改報最佳格時 DL-THR 五輪全勝且全距不重疊，
與等權口徑的「相當」結論相反（§4.2）。

**其三：量化形成期比較的一項結構性檢定力障礙**

形成期層的兩臂**必然**持有不同的配對集合，差分無法消去標的的特異變異；
交易期層的兩臂共用配對，差分即消噪。本研究量化此落差為
**標準誤相差 1.89 倍、等效樣本需求 3.6 倍**。

> 這意味著**形成期改良的實證門檻天生高於交易期改良**，
> 而既有文獻多未區分兩者的檢定力條件即並列比較其結果。
>
> 配對設計消噪的原理本身是標準統計知識；本研究的貢獻是**在配對交易的兩個
> 改良層次上把它量化出來**（1.89 倍、3.6 倍），使兩者的結果不再被等量齊觀。

配合風險評估的三項發現（獲利集中動盪期、等權口徑下成本餘裕僅 −1.8 ~ +2.4 bps、
等權組合絕對績效全數不顯著），本研究對「配對交易在 2000 年代後的美股大型股
是否仍具可提取報酬」提供**偏否定的證據**。


## 方法論貢獻（五）：產業資訊的三條管道

既有研究通常僅考慮「是否將產業納為特徵」。本研究發現產業先驗經**三條管道**進入分群：

| 管道 | 機制 | 實測影響 |
| :--- | :--- | :--- |
| **① 產業 one-hot** | 對跨產業配對施加固定距離懲罰 | 分群近鄰中跨產業距離**膨脹 77.9%**，同產業 0.0% |
| **② 缺失值插補** | 產業中位數 = 為缺資料股票加上產業標籤 | 僅改插補方式，跨產業配對 **15.7% → 59.3%** |
| **③ 排序準則** | SSD 偏好同產業（同業價格路徑天然更近） | 同分群下改用報酬發散準則，**6.8% → 19.7%** |

::: {.callout-important}

### 第二條管道是隱性的，且在多數實作中未被察覺

本研究因此發現：一項已封存的特徵消融實驗實際上是在
「**處理幾乎未施加**」的條件下執行——
原設定下加 10 個結構性特徵僅使跨產業配對 **+6.2 pp**，全域插補下為 **+42.8 pp**。

以正確設定重驗後，該實驗的原結論由「**弱 null**」升級為「**強 null**」。

:::

::: {.callout-warning}

### 範圍界定

三條管道的**強度**（77.9%、+42.8 pp、6.8%→19.7%）為本研究資料條件下的實測值，
其量級**未必外推**至其他母體或特徵集。

可外推的是方法上的提醒：**這三條管道都應逐一檢查**——
尤其第二條，它不出現在任何特徵清單裡。

:::

## 兩項本次新增的限制

::: {.callout-important}

### 其八：形成期層的檢定力不足，**且無法以更多資料補救**

兩臂標準誤相差 1.89 倍，形成期層需 **3.6 倍**樣本才能達到交易期層的解析度；
若要求 MDE 降至 0.3 pp，約需 **14 倍**樣本、**逾三個世紀**的日資料。

**因此本研究對命題 1 的 null 不具鑑別力**（已於 §4.1.3 量化）。
唯一的出路是在形成期層構造**配對設計**——使兩臂持有同一批標的而僅變動分組
——本研究未能設計出這樣的對照。

:::

::: {.callout-warning}

### 其九：與前行研究回測結果的差距未能完全解釋

本研究的回測結果與**許鈞翔（2025）**差異甚大。逐項比對其程式與本研究第三章後
共辨識出**八項**實作差異，本研究僅隔離其中**一項（進場時點）**施加受控檢定，
其餘七項——尤以交易成本假設（本研究單邊 0.29% 對其 0.10%）與報酬口徑
（$R^{CC}$ 對 $R^{EC}$）——未予檢驗。

隔離結果見**附錄 A**：改採其回歸式進場後三種排序的年化報酬**全部下降**
（−0.77 ~ −0.91 pp，BH 校正後 0/3 顯著）——即該項差異的方向**與待解釋的差距相反**。

→ 故此項不是差距的來源，但**差距本身仍未獲解釋**。

:::

::: {.aside}
附錄 A 的三條策略已計入第四章 DSR 的試驗宇宙（$N$ = 110）。
結論為負的試驗仍佔用一次試驗——試驗數不因結果不利而豁免。
:::


# 5.4 研究建議

## 對後續研究的建議

**其一：優先擴大母體，而非增加特徵**
本研究證據顯示，在 S&P 500 範圍內無論如何調整特徵集、分組方法、插補方式或
排序準則，績效皆無顯著差異；而粒度掃描顯示**候選池充足度與績效有明確關聯**。
→ 擴大至 Russell 3000 或全 CRSP 應較繼續擴充特徵維度更具邊際效益。

**其二：直接驗證槽位週轉假說**
建議實作「隨機 SKIP 且維持正常槽位配置」的交易模組並重跑回測
—— 本研究的事後重抽無法涵蓋再配置效應。

**其三：重新檢視插補方式對既有研究的影響**
產業中位數插補作為隱性管道，其影響在缺失率高時可達 **42.8 pp**。
→ 任何使用基本面特徵且缺失率不低的分群研究，結論皆可能受此影響。

**其四：報告信賴區間而非僅報告顯著與否**
方法比較若在雙尾不顯著後宣稱「表現相當」，應同時給出效果量的 CI，
並與參照臂自身的績效量級並列——區間寬度是區分「效果不存在」與
「檢定力不足」的最直接證據。

**其五：報告效果量而非僅報告顯著性**
形成期比較的效果量**結構性地**小於交易期比較（不同配對集合 vs 共用配對）
→ 僅報 $p$ 值將使讀者無法判斷「不顯著」是效果不存在還是檢定力不足。


## 對實務應用的建議

::: {.callout-important}

### 本研究**不建議**將所檢驗的策略直接應用於實務

**其一：成本餘裕過薄**
以可事前實施的等權組合計，往返 break-even 介於 **0.562% ~ 0.604%**，
較現行假設 0.58% 僅相差 **−1.8 ~ +2.4 bps**；
其中 K-means／GICS-SSD／GICS-SDP 三個 Z-Score 臂的餘裕**已經是負的**。
真實摩擦成本只要高出 1~2 bps（往返），全部配置的正報酬結論即不成立。

**其二：獲利高度集中於特定市場環境**
十組配置在平靜期**全部**為負 Sharpe（−1.24 ~ −0.20）。
策略實質上在**賣出波動率的均值回歸選擇權**，低波動環境即為結構性不利環境。

**其三：絕對績效無法與零區分**
以可事前實施的全網格等權組合計，年化 **−0.36% ~ +0.77%**，六組 CI 全部橫跨零。
附錄的 DSR 進一步顯示：即使改報最佳格，其 $SR$（最高 **0.392**）
仍**低於**純靠運氣的期望門檻（$SR_0$ = **0.408**，$N$ = 110）。

:::

**若仍欲採用，本研究的證據支持兩項設計取向**

- **形成期宜採簡單的產業分組** —— 複雜分群未顯示可辨識的優勢，卻增加實作與維護成本
- **交易期則值得投入** —— DL-THR 門檻選擇在五種配對底上皆一致改善，且具 regime 穩健性


# 5.5 結語

## 過程所揭露的三件事

兩個命題得到方向相反的答覆：交易期的**門檻選擇**改良獲得支持，
形成期的**機器學習分組**改良則否。但更值得記述的或許不是答覆本身，
而是取得答覆的過程所揭露的三件事。

::: {.callout-important}

**其一：推論的正當性取決於抽樣單位的選擇**

同一批資料、同一個效果量，在錯誤的抽樣單位下可以得到 $p < 0.001$；
在正確的抽樣單位下，則需要完全不同的檢定方法才能建立。

**其二：「未獲支持」與「無效」不是同一件事**

本研究對命題 1 施加三輪受控消融、逐一排除實作原因，
才能將結論從「我們的實作沒做出效果」提升為
「**在此設計與樣本下，形成期搜尋空間的改良無法被驗證**」。

這仍不是「它沒有效果」——**區分這三種說法，是本研究最花力氣的一件事。**

**其三：設計選擇會以未預期的方式攜帶資訊**

產業先驗不只存在於明確的 one-hot 編碼，
也潛藏於**缺失值的插補規則**與**距離準則的選擇**之中。
若不逐一檢視，研究者可能在自以為移除某項先驗時，實際上仍保留了它。

:::
