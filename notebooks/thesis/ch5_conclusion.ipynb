# 5.1 研究發現綜整

## 兩個命題，方向相反的答覆

| | 命題 1（形成期） | 命題 2（交易期） |
| :--- | :--- | :--- |
| **主張** | ML 分群能建立更優的配對搜尋空間 | DRL 交易端優於 Z-Score 規則 |
| **結果** | **未獲支持** | **獲得支持** |
| **證據** | 9 組對照 BH 後無一顯著（最小 $p$=0.377），方向 8/9 偏 GICS | 五配對底全部顯著（$p$=0.0000–0.0042） |
| **信賴區間** | 9 組**全部涵蓋 0**，寬 1.2–1.9 pp | 五組**全部落在零的右側**，最保守下界 +0.22 pp |
| **穩健性** | HAC 對照一致 | HAC 對照一致 |

::: {.callout-important}

### 命題 1 是「未能拒絕虛無假設」，**非**「GICS 顯著較優」

9 組區間的寬度（1.2–1.9 pp）是 GICS 參照臂**自身**年化績效
（−0.11% ~ −0.40%）的**數倍**。

→ **本研究對此比較檢定力不足**：既無法證明 ML 較優，亦無法主張兩者相當。

:::

## 組合系統：未達顯著，且改善只來自一半

標題宣稱的比較——動態分群 + DRL **vs** GICS + Z-Score：

| 分群法 | 總效果 | DRL 成分 | 分群成分 |
| :--- | ---: | ---: | ---: |
| Agglomerative | +0.818 | **+0.719** ✔ | +0.099 ✘ |
| HDBSCAN | +0.715 | **+0.896** ✔ | −0.182 ✘ |
| K-means | +0.483 | **+0.589** ✔ | −0.106 ✘ |

- 總效果方向一致為正，**BH 校正後無一達 5% 顯著**
- **全部改善來自 DRL 成分**；分群成分三者中兩者為負

> 本研究所提策略的價值集中於**交易端**。
> 無監督動態分群在此資料條件下未對組合系統產生正向貢獻。

## 命題 2 的增益機制已釐清

**三項替代解釋皆被排除**

| 替代解釋 | 檢驗方式 | 結果 |
| :--- | :--- | :--- |
| 拉高進場門檻 | 同門檻（2.2）對照 | 五支仍顯著；門檻管道僅複製 **4.2–20.9%**，無一顯著 |
| 篩掉劣質配對 | SKIP 置換檢定 | **75 格僅 7 格顯著**（隨機期望 3.75） |
| 減少曝險 | 進場次數比較 | DRL 反為 Z-Score 的 **1.6–1.9 倍** |

殘差與「**槽位週轉**」假說一致 —— 惟需重跑回測方能直接驗證。

## 命題 1 的失敗不可歸因於實作選擇

三項候選原因逐一受控檢驗，**皆使形成期結構如預期改變，但績效皆未改善**：

1. **形成期設計** —— 產業 one-hot、共整合篩選、缺「不分組」零點
2. **交易機制** —— Han et al. 四項差異逐步施加
3. **特徵維度** —— 連續特徵 7 → 19 維

## 核心發現與絕對績效

::: {.callout-important}

### 選哪些配對，不決定報酬

六種設定涵蓋 **one-hot 權重、插補方式、特徵集、分組方法**四個維度，
跨產業配對比例自 **0%（GICS）變動至 69%（完全不分組）**，
而 2012 年後績效全部落在 **−1.8% ~ −0.7%、彼此無顯著差異**。

:::

**絕對獲利能力未獲證實**

以可事前實施的**全網格等權組合**計，六組配置年化落在 **−0.44% ~ +0.61%**，
bootstrap 檢定**無一顯著**，信賴區間寬達 **2 ~ 3.5 pp**。

> 此非樣本不足而是**結構性限制**：等權組合年化 Sharpe 僅 **0.05 ~ 0.18**，
> 在此量級下欲達統計顯著約需 **123 年以上**樣本。

# 5.2 研究貢獻

## 方法論貢獻（一）：抽樣單位的偽重複

配對交易文獻常以「參數網格」為觀測單位進行跨策略檢定。

::: {.callout-important}

### 這構成偽重複（pseudo-replication）

15 個參數配置共用**同一份資料、同一段期間、同一批配對**：
`top_n`=10 與 20 共用 10 組配對；三種停損是同一批交易的不同出場規則。

→ 有效樣本數接近**一條回測路徑**，而非 15。

:::

改以時間為抽樣單位後：

| | 舊版（參數網格） | 新版（時間維度） |
| :--- | ---: | ---: |
| 命題 2 效果量 | +0.732 pp | **+0.719 pp** |
| 推論正當性 | ✘ 獨立性假設不成立 | ✔ block bootstrap（$p$ 值 + 95% CI） |

**效果量幾乎不變，但推論的正當性完全不同。**
此區別對任何以參數網格作跨策略比較的研究皆適用。

## 方法論貢獻（二）：中性組裝器

將形成期拆解為四個可獨立替換的層：

```
特徵萃取  →  分組  →  群內排序  →  統計篩選
```

使「任兩策略之間僅存單一變因」成為**架構性質**，而非人為約定。

::: {.callout-tip}

此設計使本研究得以在命題未獲支持時，**系統性地逐一排除實作原因**
—— 而非僅能宣稱「方法無效」。

:::

## 方法論貢獻（四）：區分「未獲支持」與「兩者相當」

雙尾檢定不顯著時，文獻常逕行宣稱兩方法「表現相當」。

本研究主張此判斷應交由**信賴區間的寬度**承擔：

- 區間涵蓋 0 且**兩端皆小** → 方可討論「相當」
- 區間涵蓋 0 但**兩端皆大於參照臂自身的績效量級** → 只能說**檢定力不足**

此作法不需預先指定容忍邊界，避免了非劣性檢定中
「$\delta$ 取多少」這一難以辯護的判斷。

## 方法論貢獻（三）：產業資訊的三條管道

既有研究通常僅考慮「是否將產業納為特徵」。本研究發現產業先驗經**三條管道**進入分群：

| 管道 | 機制 | 實測影響 |
| :--- | :--- | :--- |
| **① 產業 one-hot** | 對跨產業配對施加固定距離懲罰 | 分群近鄰中跨產業距離**膨脹 77.9%**，同產業 0.0% |
| **② 缺失值插補** | 產業中位數 = 為缺資料股票加上產業標籤 | 僅改插補方式，跨產業配對 **15.7% → 59.3%** |
| **③ 排序準則** | SSD 偏好同產業（同業價格路徑天然更近） | 同分群下改用報酬發散準則，**6.8% → 19.7%** |

::: {.callout-important}

### 第二條管道是隱性的，且在多數實作中未被察覺

本研究因此發現：一項已封存的特徵消融實驗實際上是在
「**處理幾乎未施加**」的條件下執行——
原設定下加 10 個結構性特徵僅使跨產業配對 **+6.2 pp**，全域插補下為 **+42.8 pp**。

以正確設定重驗後，該實驗的原結論由「**弱 null**」升級為「**強 null**」。

:::

## 實證貢獻

**其一：為機器學習分群的適用邊界提供反面證據**

Han et al. (2021) 以 CRSP 全美股、126 維特徵報告年化 **24.8%**。
本研究在 S&P 500、可得特徵子集下未能複製，且受控消融排除三項實作原因。

> 此結果**不否定**原研究，而是界定其**適用範圍**：
> 該方法的優勢可能依賴**母體廣度**與**特徵維度**，受限資料條件下不必然成立。

**其二：釐清 DRL 增益的來源與非來源**

排除三項最直觀的替代解釋 → 「DRL 確有貢獻」建立在更窄、也更可信的基礎上。
特別指出：損益分解中的「SKIP 貢獻 31–52%」是**會計恆等式而非技巧歸因**。

**其三：提出報酬來源的結構性觀察**

「選哪些配對」在極大範圍內不影響報酬。配合三項風險發現
（獲利集中動盪期、成本餘裕 6.5–12.8 bps、等權組合絕對績效全數不顯著），
對「配對交易在 2000 年代後的美股大型股是否仍具可提取報酬」提供**偏否定的證據**。

| # | 限制 | 偏誤方向 |

# 5.4 研究建議

## 對後續研究的建議

**其一：優先擴大母體，而非增加特徵**
本研究證據顯示，在 S&P 500 範圍內無論如何調整特徵集、分組方法、插補方式或
排序準則，績效皆無顯著差異；而粒度掃描顯示**候選池充足度與績效有明確關聯**。
→ 擴大至 Russell 3000 或全 CRSP 應較繼續擴充特徵維度更具邊際效益。

**其二：直接驗證槽位週轉假說**
建議實作「隨機 SKIP 且維持正常槽位配置」的交易模組並重跑回測
—— 本研究的事後重抽無法涵蓋再配置效應。

**其三：重新檢視插補方式對既有研究的影響**
產業中位數插補作為隱性管道，其影響在缺失率高時可達 **42.8 pp**。
→ 任何使用基本面特徵且缺失率不低的分群研究，結論皆可能受此影響。

**其四：報告信賴區間而非僅報告顯著與否**
方法比較若在雙尾不顯著後宣稱「表現相當」，應同時給出效果量的 CI，
並與參照臂自身的績效量級並列——區間寬度是區分「效果不存在」與
「檢定力不足」的最直接證據。

**其五：報告效果量而非僅報告顯著性**
形成期比較的效果量**結構性地**小於交易期比較（不同配對集合 vs 共用配對）
→ 僅報 $p$ 值將使讀者無法判斷「不顯著」是效果不存在還是檢定力不足。

## 對實務應用的建議

::: {.callout-important}

### 本研究**不建議**將所檢驗的策略直接應用於實務

**其一：成本餘裕過薄**
最佳策略往返 break-even 為 0.708%，較現行假設 0.58% 僅高 **12.8 bps**。
真實摩擦成本若較假設高出 4 bps（單邊），多數策略的正報酬結論即不成立。

**其二：獲利高度集中於特定市場環境**
所有策略在平靜期皆為**負 Sharpe**（−0.72 ~ −0.05）。
策略實質上在**賣出波動率的均值回歸選擇權**，低波動環境即為結構性不利環境。

**其三：絕對績效無法與零區分**
以可事前實施的全網格等權組合計，年化 −0.44% ~ +0.61%，六組 CI 全部橫跨零。
附錄的 DSR 進一步顯示：即使改報最佳格，其 $SR$（最高 0.343）仍低於
純靠運氣的期望門檻（0.415）。

:::

**若仍欲採用，本研究的證據支持兩項設計取向**

- **形成期宜採簡單的產業分組** —— 複雜分群未顯示可辨識的優勢，卻增加實作與維護成本
- **交易期則值得投入** —— DRL 門檻選擇在五種配對底上皆一致改善，且具 regime 穩健性

# 5.5 結語

## 過程所揭露的三件事

兩個命題得到方向相反的答覆。但更值得記述的或許不是答覆本身，
而是取得答覆的過程所揭露的三件事。

::: {.callout-important}

**其一：推論的正當性取決於抽樣單位的選擇**

同一批資料、同一個效果量，在錯誤的抽樣單位下可以得到 $p < 0.001$；
在正確的抽樣單位下，則需要完全不同的檢定方法才能建立。

**其二：「未獲支持」與「無效」不是同一件事**

本研究對命題 1 施加三輪受控消融、逐一排除實作原因，
才能將結論從「我們的實作沒做出效果」提升為
「**在此資料條件下，形成期搜尋空間設計的邊際貢獻趨近於零**」。

**其三：設計選擇會以未預期的方式攜帶資訊**

產業先驗不只存在於明確的 one-hot 編碼，
也潛藏於**缺失值的插補規則**與**距離準則的選擇**之中。
若不逐一檢視，研究者可能在自以為移除某項先驗時，實際上仍保留了它。

:::
