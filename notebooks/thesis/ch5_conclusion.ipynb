# 5.1 研究發現綜整

## 三項檢定，校正後皆無顯著

本研究將配對交易的改良拆解為**形成期的分組方法**與**交易期的門檻選擇**兩層，
分別檢驗其貢獻，並在**系統層次**檢驗兩者的組合。

| 檢定 | 內容 | 方向 | 校正後顯著 |
| :--- | :--- | :--- | :---: |
| 分組層 | 資料驅動分群 vs GICS 產業分組 | **9/9 偏向 GICS** | 0/9 |
| 交易層 | DL-THR vs 固定門檻 Z-Score | **5/5 為正** | 0/5 |
| 兩層組合 | 完整系統 vs 傳統基準 | **全期 3/3 為負** | 0/3 |

::: {.callout-important}
**方向的一致性是本研究最可靠的訊號，而非任何單組的 $p$ 值。**
:::

> **組合系統**（資料驅動分群 ＋ DL-THR）**劣於**傳統基準（GICS ＋ 固定門檻）
> ——全期三組年化Δ 為 −0.265 ~ −0.630 pp。原因見下頁。

## 成分分解：損害來自哪一半？

$$(\text{分群}+\text{DL-THR}) - (\text{GICS}+\text{ZS}) = \underbrace{\cdots}_{\text{DL-THR 成分}} + \underbrace{\cdots}_{\text{分群成分}}$$

| 期間 | 分群法 | 總效果 | DL-THR 成分 | $p$ | 分群成分 | $p$ |
| :--- | :--- | ---: | ---: | ---: | ---: | ---: |
| 全期 | Agglomerative | −0.334 | **+0.112** | 0.488 | **−0.447** | 0.322 |
| 全期 | HDBSCAN | −0.265 | **+0.456** | 0.052 | **−0.721** | 0.061 |
| 全期 | K-means | −0.630 | **+0.283** | 0.073 | **−0.913** | 0.080 |
| 2012+ | Agglomerative | +0.470 | +0.279 | 0.064 | +0.191 | 0.590 |
| 2012+ | HDBSCAN | +0.193 | +0.511 | 0.013 | −0.318 | 0.387 |
| 2012+ | K-means | +0.582 | +0.424 | 0.009 | +0.158 | 0.687 |

::: {.callout-important}
**全期三組的 DL-THR 成分皆正、分群成分皆負，且分群成分的量級更大。**

完整系統之所以劣於傳統基準，是因為**分群層的損害超過交易端的貢獻**。
兩個成分並不互補。
:::

> 若要保留交易端的改良，應搭配**傳統產業分組或不分組**，而非資料驅動分群。

## 分層歸因：兩層的方向相反

| | 形成期分組層 | 交易期門檻選擇 |
| :--- | :--- | :--- |
| **待檢問題** | 分組施加的限制是否提升配對品質 | 逐期選擇門檻是否優於固定門檻 |
| **方向** | **9/9 偏向 GICS**（限制愈強愈差） | **5/5 為正** |
| **效果量** | −0.229 ~ −1.000 pp | +0.112 ~ +0.573 pp |
| **校正後顯著** | 0/9 | 0/5 |
| **MDE** | 1.259 pp | 0.624 pp |
| **不顯著的性質** | MDE **大於**參照臂自身年化 → null 幾乎不帶資訊 | 效果與 MDE 同量級 → 接近偵測極限 |

::: {.callout-important}
**兩者皆不顯著，但性質截然不同。**

分組層：資料無法排除「資料驅動分群顯著較差」，其程度遠甚於無法排除「兩者相當」。

交易層：效果存在但量級接近偵測極限。
:::

> 納入「不分組」零點後型態更完整：
> **不分組 > GICS > 任一資料驅動分群**，15 格逐格成立。

## 交易層的增益機制

**三項機械性替代解釋，在兩條 GICS 底皆被排除**

| 替代解釋 | 檢驗方式 | 結果 |
| :--- | :--- | :--- |
| SKIP 的選股方向 | 置換檢定（2,000 次／格） | **75 格僅 4 格顯著**（隨機期望 3.75）；AGG 與 KM 的技巧成分為**負** |
| 門檻水準 | 拉到同門檻（2.1/2.2/2.3）對照 | 門檻管道僅複製 **0–30%** |
| 總曝險減少 | 固定門檻 + 隨機跳過同數量 | GICS 兩底 **11/15、10/15 格顯著**勝出 |

| 配對底 | SKIP 選股 | 門檻水準 | 總曝險 | 判定 |
| :--- | :---: | :---: | :---: | :--- |
| GICS-SSD | 排除 | 排除 | **排除** | 三項全排除 |
| GICS-SDP | 排除 | 排除 | **排除** | 三項全排除 |
| HDBSCAN | 排除 | 排除 | 部分 | 證據較弱 |
| K-means | **否證** | 排除 | 未排除 | 增益屬機械效應 |
| Agglomerative | **否證** | — | **未排除** | 無增益 |

::: {.callout-warning}
三項檢定皆為**事後重抽，不含槽位再配置效應**——
回答的是三項機械管道是否構成解釋，而非這些動作整體有無價值。
:::

## 核心發現：兩層的檢定力不對等來自**設計**

::: {.callout-important}

### 不對等的成因

交易層的兩臂**共用**同一批配對，逐日差分消去共同的市場衝擊；
分組層的兩臂持有**不同**的配對集合，差分中混入兩套標的的特異變異。

標準誤相差 **2.02 倍** → **等效於形成期層需 4.1 倍的樣本期間**。
若要求 MDE 降至 0.3 pp，形成期層約需 **17.6 倍**樣本、**逾四個世紀**的日資料。

:::

| 層 | $SE$ 中位 | MDE | 對 0.3 pp 的檢定力 |
| :--- | ---: | ---: | ---: |
| 形成期（分組層） | 0.449 | **1.259 pp** | **10%** |
| 交易期（交易層） | 0.223 | **0.624 pp** | 27% |

> **此為結構性限制**：更換分群演算法、增加特徵維度、改良插補方式皆不影響它。
>
> 唯一的出路是在形成期層構造**配對設計**——使兩臂持有同一批標的而僅變動分組。
> 本研究未能設計出這樣的對照，列為後續研究的首要方法論課題。

# 5.2 研究貢獻

::: {.callout-warning}

### 先界定本節的口氣

以下三項**皆非新方法**。偽重複、單變因設計、信賴區間的解讀，
在統計與實驗設計文獻中都是既有的標準作法。

本研究的貢獻在於**把它們搬進配對交易的比較研究裡並貫徹到底**——
該領域的慣例與這些標準之間存在明確落差，而本研究記錄了落差的後果。

:::

## 方法論貢獻（一）：把偽重複問題帶進配對交易的跨策略檢定

配對交易文獻常以「參數網格」為觀測單位進行跨策略檢定。

> **偽重複（pseudo-replication）** 一詞出自 **Hurlbert (1984)**,
> *Pseudoreplication and the design of ecological field experiments*,
> **Ecological Monographs, 54(2), 187–211** —— 指**當多個「觀測」其實共用同一個
> 實驗單元時，把它們當成獨立樣本會虛增自由度**。該概念在生態學已是標準批評，
> 但**未見於配對交易文獻的檢定設計討論**。

本研究的 15 個參數配置正是此情形：共用同一份資料、同一段期間、同一批配對
（`top_n`=10 與 20 共用 10 組配對；三種停損是同一批交易的不同出場規則）
→ 有效樣本數接近**一條回測路徑**，而非 15。

| | 以參數網格為單位 | 以時間為單位 |
| :--- | ---: | ---: |
| 交易層 效果量 | +0.732 pp | **+0.787 pp** |
| 推論正當性 | ✘ 獨立性假設不成立 | ✔ block bootstrap（$p$ 值 + 95% CI） |

**效果量幾乎不變，但推論的正當性完全不同。**

> **可主張者**：本研究記錄了「以參數網格為觀測單位」在配對交易脈絡下的具體後果，
> 並提供一個可照做的替代口徑。
> **不可主張者**：本研究「發現」了偽重複問題。


## 方法論貢獻（一）續：以區間寬度區分兩種「不顯著」

雙尾檢定不顯著時，文獻常逕行宣稱兩方法「表現相當」。

> 此一謬誤有明確的方法學來源可引：
> **Altman & Bland (1995)**, *Absence of evidence is not evidence of absence*,
> **BMJ, 311(7003), 485**；
> **Greenland et al. (2016)**, *Statistical tests, P values, confidence intervals,
> and power: a guide to misinterpretations*,
> **European Journal of Epidemiology, 31(4), 337–350**；
> **Wasserstein & Lazar (2016)**, *The ASA statement on p-values*,
> **The American Statistician, 70(2), 129–133**。

本研究據此主張：該判斷應交由**信賴區間的寬度**承擔。

| 情形 | 可下的結論 |
| :--- | :--- |
| 區間涵蓋 0，**兩端皆小**（相對於參照臂自身的績效量級） | 可討論「**相當**」 |
| 區間涵蓋 0，**兩端皆大** | 只能說**檢定力不足** |

::: {.callout-note}

### 與等價檢定的關係

正式作法是**等價檢定**（equivalence testing，如 **Lakens, 2017**,
*Equivalence tests*, **Social Psychological and Personality Science, 8(4), 355–362**），
但它要求事前指定容忍邊界 $\delta$，而配對交易並無公認的 $\delta$。

本研究因此**不宣稱做了等價檢定**，僅以區間寬度作定性區分——
這是較弱、但不需要辯護 $\delta$ 的作法。
故第四章對「配對選擇不影響報酬」一類主張一律不予採納（§4.5.2）。

:::


## 方法論貢獻（二）：把單變因原則實作為架構保證

**單變因對照本身是實驗設計的基本原則**（一次只變動一個因子，
可上溯至 **Fisher, 1935**, *The Design of Experiments*）。
本研究的貢獻不在原則，而在**把它做成程式架構的性質**。

將形成期拆解為四個可獨立替換的層：

```
特徵萃取  →  分組  →  群內排序  →  統計篩選
```

各層以標準化介面銜接，任兩策略之組態差異可被機械地檢查。

::: {.callout-important}

### 為何這件事值得記述

配對交易文獻中，「更換分群方法」的實作通常伴隨其他環節一併改動
（不同的排序準則、不同的篩選門檻），使績效差異**無法歸因至任一單項**。
本研究第二章已將此列為缺口之一。

把單變因做成**架構保證**而非人為約定後，本研究得以在某項改良未顯示效果時
**系統性地逐一排除實作原因**——形成期設計、交易機制、特徵維度三項候選
全部受控檢驗且皆非原因（§4.4）——而不只是宣稱「方法無效」。

:::

> **可主張者**：一個使單變因對照可被機械驗證的實作架構。
> **不可主張者**：單變因設計是本研究提出的。

::: {.callout-warning}

### 這項保證也有一處例外，已於第三章揭露

排序層的替換同時改變了排序與篩選的**施行次序**，
故排序層之間的比較嚴格說並非單一變因（§3.2）。
本研究的兩項主要檢定落在分組層與交易端，皆固定排序後端，不受此影響。

:::


## 方法論貢獻（三）：限制的量化揭露與偏誤方向

配對交易文獻的限制章節多為定性陳述（「樣本有限」「成本假設可能低估」），
讀者無從判斷某項限制會使結論**偏哪一邊**、偏多少。

> 此一要求同樣有來源：**Harvey (2017)**, *Presidential address: The scientific
> outlook in financial economics*, **Journal of Finance, 72(4), 1399–1440**
> 對財務實證研究的可信度提出系統性批評，主張研究者應揭露完整的試驗歷程
> 與偏誤方向，而非僅呈現最終結果。

本研究對每項限制報告**具體數字**與**偏誤方向**：

| 限制 | 量化 | 偏誤方向 |
| :--- | :--- | :--- |
| 存活者偏誤 | 170 檔已下市股中 49 檔（29%）無價格，且缺失與下市狀態相關（43% vs 17%） | 絕對績效**被高估** → 「不顯著」結論**偏保守** |
| 成本假設年代 | 0.29%/邊取自 Do & Faff (2012) 對 1962–2009 的估計，套用於 2000–2025 | 摩擦成本長期下降 → **偏保守**；另報 break-even 使結論不依賴此假設 |
| 特徵覆蓋率 | 自 SEC XBRL 建出 40 個，僅 12 個通過 70% 門檻；另 5 個評價類卡在 57–70%、11 個永遠無法取得 | 分組層的特徵維度結論僅在「SEC 申報所能供給的範圍內」成立 |

::: {.callout-important}

### 方向已知的限制，不對稱地支持結論

三項主要限制**全部**指向「絕對績效被高估」。
本研究的絕對績效結論是「無一顯著」——限制若被修正，
該結論只會更強而不會翻轉。

反之，分組層的「未獲支持」則受母體範圍限制影響，**方向不明**，
故明確限定於本設定。

:::

> 本研究另將試驗宇宙實地清點並釘死（$N$ = 110，§4.6.3），
> 使 Deflated Sharpe（**Bailey & López de Prado, 2014**）的關鍵參數可被檢查，
> 而非由研究者事後選定。


## 實證貢獻

**其一：為資料驅動分群的適用邊界提供反面觀察**
Han, He & Toh (2021) 以 CRSP 全美股報告年化 24.8%；本研究在受限資料條件下未能複製。
惟分組層的檢定力不足（MDE **1.259 pp**），此觀察本身的強度有限。

**其二：釐清門檻選擇增益的來源與非來源**
以三個虛無分布逐一檢驗（SKIP 選股、門檻水準、總曝險）。
指出「SKIP 貢獻」是**會計恆等式而非技巧歸因**——置換檢定證實兩個配對底的
技巧成分為**負**。

**其三：一項被自身資料否證的設計假設**
全資訊性質原被預期為樣本效率之來源。以逐位元相同、僅將標籤縮為部分回饋的
RL-THR 對照後，**三個變體點估計皆優於全資訊版本**（+0.158~+0.217 pp，不顯著）。
→ **資料不支持「全資訊是效果之來源」**。

**其四：指出形成期比較的結構性檢定力障礙**
量化為標準誤相差 **2.02 倍**、等效樣本需求 **4.1 倍**。

**其五：記錄一組實作錯誤及其對結論的影響**
六項實作與宣告不符（附錄 B），修正後主軸 13 格中 12 格改善，$\Delta$Sharpe 中位 **+0.236**。

## 方法論貢獻（五）：產業資訊的三條管道

既有研究通常僅考慮「是否將產業納為特徵」。本研究發現產業先驗經**三條管道**進入分群：

| 管道 | 機制 | 實測影響 |
| :--- | :--- | :--- |
| **① 產業 one-hot** | 對跨產業配對施加固定距離懲罰 | 分群近鄰中跨產業距離**膨脹 77.9%**，同產業 0.0% |
| **② 缺失值插補** | 產業中位數 = 為缺資料股票加上產業標籤 | 僅改插補方式，跨產業配對 **15.7% → 59.3%** |
| **③ 排序準則** | SSD 偏好同產業（同業價格路徑天然更近） | 同分群下改用報酬發散準則，**6.8% → 19.7%** |

::: {.callout-important}

### 第二條管道是隱性的，且在多數實作中未被察覺

本研究因此發現：一項已封存的特徵消融實驗實際上是在
「**處理幾乎未施加**」的條件下執行——
原設定下加 10 個結構性特徵僅使跨產業配對 **+6.2 pp**，全域插補下為 **+42.8 pp**。

以正確設定重驗後，該實驗的原結論由「**弱 null**」升級為「**強 null**」。

:::

::: {.callout-warning}

### 範圍界定

三條管道的**強度**（77.9%、+42.8 pp、6.8%→19.7%）為本研究資料條件下的實測值，
其量級**未必外推**至其他母體或特徵集。

可外推的是方法上的提醒：**這三條管道都應逐一檢查**——
尤其第二條，它不出現在任何特徵清單裡。

:::

## 兩項本次新增的限制

::: {.callout-important}

### 其八：形成期層的檢定力不足，**且無法以更多資料補救**

兩臂標準誤相差 **2.02 倍**，形成期層需 **4.1 倍**樣本才能達到交易期層的解析度；
若要求 MDE 降至 0.3 pp，約需 **17.6 倍**樣本、**逾四個世紀**的日資料。

**因此本研究對分組層的 null 不具鑑別力**（已於 §4.5 量化）。
唯一的出路是在形成期層構造**配對設計**——使兩臂持有同一批標的而僅變動分組
——本研究未能設計出這樣的對照。

:::

::: {.callout-warning}

### 其九：與前行研究回測結果的差距未能完全解釋

逐項比對後辨識出八項實作差異，本研究僅隔離其中一項（進場時點）作受控檢定，
其餘七項未予檢驗。就量級而言，**成本假設**（0.29% 對 0.10%）
與**報酬口徑**（$R^{CC}$ 對 $R^{EC}$，本研究兩者相差 5–6 倍）
仍是最可能的主導因素。詳見附錄 A。

:::

# 5.4 研究建議

## 對後續研究的建議

**其一：優先擴大母體，而非增加特徵**
本研究證據顯示，在 S&P 500 範圍內無論如何調整特徵集、分組方法、插補方式或
排序準則，績效皆無顯著差異；而粒度掃描顯示**候選池充足度與績效有明確關聯**。
→ 擴大至 Russell 3000 或全 CRSP 應較繼續擴充特徵維度更具邊際效益。

**其二：直接驗證槽位週轉假說**
建議實作「隨機 SKIP 且維持正常槽位配置」的交易模組並重跑回測
—— 本研究的事後重抽無法涵蓋再配置效應。

**其三：重新檢視插補方式對既有研究的影響**
產業中位數插補作為隱性管道，其影響在缺失率高時可達 **42.8 pp**。
→ 任何使用基本面特徵且缺失率不低的分群研究，結論皆可能受此影響。

**其四：報告信賴區間而非僅報告顯著與否**
方法比較若在雙尾不顯著後宣稱「表現相當」，應同時給出效果量的 CI，
並與參照臂自身的績效量級並列——區間寬度是區分「效果不存在」與
「檢定力不足」的最直接證據。

**其五：報告效果量而非僅報告顯著性**
形成期比較的效果量**結構性地**小於交易期比較（不同配對集合 vs 共用配對）
→ 僅報 $p$ 值將使讀者無法判斷「不顯著」是效果不存在還是檢定力不足。


## 對實務應用的建議

::: {.callout-important}

### 本研究**不建議**將所檢驗的策略直接應用於實務

**其一：成本餘裕過薄**
以可事前實施的等權組合計，往返 break-even 介於 **0.562% ~ 0.604%**，
較現行假設 0.58% 僅相差 **−1.8 ~ +2.4 bps**；
其中 K-means／GICS-SSD／GICS-SDP 三個 Z-Score 臂的餘裕**已經是負的**。
真實摩擦成本只要高出 1~2 bps（往返），全部配置的正報酬結論即不成立。

**其二：獲利高度集中於特定市場環境**
十組配置在平靜期**全部**為負 Sharpe（−1.24 ~ −0.20）。
策略實質上在**賣出波動率的均值回歸選擇權**，低波動環境即為結構性不利環境。

**其三：絕對績效無法與零區分**
以可事前實施的全網格等權組合計，年化 **−0.36% ~ +0.77%**，六組 CI 全部橫跨零。
附錄的 DSR 進一步顯示：即使改報最佳格，其 $SR$（最高 **0.392**）
仍**低於**純靠運氣的期望門檻（$SR_0$ = **0.328**，$N$ = 44）。

:::

**若仍欲採用，本研究的證據支持兩項設計取向**

- **形成期宜採簡單的產業分組** —— 複雜分群未顯示可辨識的優勢，卻增加實作與維護成本
- **交易期則值得投入** —— DL-THR 門檻選擇在五種配對底上皆一致改善，且具 regime 穩健性


# 5.5 結語

## 過程所揭露的四件事

三項檢定經多重檢定校正後**皆未獲統計支持**，但方向並不對稱：
交易期的門檻選擇一致為正，形成期的分組改良一致為負，兩者組合後淨效果亦為負。
更值得記述的或許不是這些答覆本身，而是取得答覆的過程所揭露的四件事。

::: {.callout-important}

**其一：推論的正當性取決於抽樣單位的選擇**

同一批資料、同一個效果量，在錯誤的抽樣單位下可以得到 $p < 0.001$；
在正確的抽樣單位下，則需要完全不同的檢定方法才能建立。

**其二：「未獲支持」與「無效」不是同一件事**

本研究對分組層施加三輪受控消融、逐一排除實作原因，
才能將結論從「我們的實作沒做出效果」提升為
「在此設計與樣本下，該項改良無法被驗證」。這仍不是「它沒有效果」。

**其三：設計選擇會以未預期的方式攜帶資訊**

產業先驗不只存在於明確的 one-hot 編碼，也潛藏於缺失值的插補規則與
距離準則的選擇之中。研究者可能在自以為移除某項先驗時，實際上仍保留了它。

**其四：實作與宣告之間的落差可以大到改變結論的方向**

六項形成期實作與方法章節所述不符——共整合檢定誤用了另一個分布的臨界值、
均值回歸指標測量的是錯誤的對象、密度分群的噪音標記被當成一個群。
修正後 GICS 三格自負 Sharpe 全數翻正。

**方法章節寫了什麼，與程式實際做了什麼，是兩件需要分別驗證的事。**

:::
