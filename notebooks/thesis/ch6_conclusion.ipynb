# 6.1 研究發現綜整

以 S&P 500 成分股 2000–2025 年日頻資料，在受控的單變因設計下
檢驗**動作空間的設計對學習式配對交易端的影響**。

| 段 | 問題 | 答案 | 證據等級 |
| :--- | :--- | :--- | :--- |
| 一（4.1） | 逐日自由持倉是否系統性造成過度交易與劣化 | **是** | 完整預先登記 |
| 二（4.2） | 縮小動作空間後，逐對辨識能力是否存在 | **否** | 跑前定稿 |
| 三（4.3） | 相對固定門檻是否有效 | **是**（但不可交易） | 分析時判準 |

> **三個答案並不一致，而這種不一致本身就是本研究的發現。**

## 其一，動作空間的設計是關鍵約束

在配對來源、對沖口徑、網格格點與訓練預算**完全相同**的條件下，
三代逐日自由持倉（online DQN、其修復版、FQI）的換手率為門檻選擇式的
**2.2–4.1 倍**，三條臂全部同時贏過兩個基準
（BH 校正 $p \le 0.0004$，達標 3/3，判準僅要求 2/3），Sharpe 一律不高於門檻選擇式。

結論在兩種進場定義下皆成立（比值 2.868–4.095 對 2.849–4.067，BH 校正 $p$ 完全相同）。

> **表達能力更強的動作空間在日頻上表現更差**——
> 這與「逐日決策可模擬任何固定門檻規則，故應弱優於後者」的直覺相反。

**成本歸因顯示三代的失敗原因並不相同**：v1／v2 的劣勢主要是換手成本
（零成本後差距縮小 87.0%／83.4%），v3（FQI）不是（僅 31.7%；
零成本下 ΔSharpe 由 −0.285 轉為 **+0.195**，成本一加回就翻負）。

## 其二，增益不來自逐對辨識能力

把動作空間縮到「每期選一組門檻」後，以監督回歸逐對學門檻：

| | 每期平均 `capture_frac` |
| :--- | ---: |
| 模型 | +0.005510 |
| **擴張窗最佳常數** | **+0.010397** |

配對 $t$，$p$ = **0.0004**——模型不僅未捕獲，且**顯著劣於一個常數**
（吃到完美預知天花板的 **−13.8%**）。

**機制屬資訊科學而非財務推論**：依據的訊號 `Spearman` = +0.1007（弱），
而選取方式是對 20 個獨立回歸取 **argmax** → 弱訊號下 argmax 系統性偏向
預測誤差最大的輸出，**選取本身引入的雜訊超過訊號**。

> 與自由持倉 DRL 的失敗是同一件事的兩面：後者塌陷至邊際平均解
> （門檻中位 2.11–2.33、`Spearman(門檻, 名次)` ≈ 0），前者則**比常數更差**。
> **逐對門檻選擇所需的訊號，在現有特徵中不足。**

## 其三，形狀確有價值，但不足以構成可交易的策略

門檻選擇式相對固定門檻 Z-Score，五個配對來源方向**全部為正**
（年化 +0.239 ~ +0.798 pp），**BH-FDR 校正後 3/5 顯著**；
增益最大者為**傳統 GICS 來源**，故其價值與資料驅動分群無關。
三項機械性替代解釋在 GICS-SSD、HDBSCAN、GICS-SDP 上全部排除。

> **但它跨不過可交易的門檻**，且是三道獨立界線同時不過（5.1 節）：
>
> - 絕對績效最佳配置帳面年化 **1.026%**，閒置現金計息後的 rf 超額僅
>   **+0.375 個百分點**（Sharpe 0.276 < DSR 門檻 0.408），
>   且對「日均報酬是否為零」的檢定 **6/6 不顯著**。該口徑下
>   **15 格固定門檻基準無一為正**（最高 +0.009 pp），而 DL-THR 的超額中
>   有 **+0.282 pp 只是少動用資本多收的利息**，不應計為技巧
> - **Deflated Sharpe Ratio 41 條策略無一通過 0.95**
> - 成本餘裕全距 **58 bps**，與成本假設本身同量級，其中 K-means 為負

## 其四，三項結果合起來的一句話

> # 門檻選擇的增益來自動作空間的**設計**，不來自逐對**辨識能力**；
> # 而該增益的量級**不足以構成可交易的策略**。

## 其五，兩項附帶但值得記述的否證

**其一，全資訊性質被自己設置的對照組否證。**
本研究交易端的 9 個動作報酬皆可反事實回算，原被預期為其樣本效率之來源；
以逐位元相同、僅將訓練標籤縮為部分回饋的 RL-THR 作對照後，
**三個部分回饋變體的點估計皆優於全資訊版本**（差異不顯著）。

**其二，文獻成功案例的頻率相依。**
AI 於配對交易的成功案例，其資料頻率高度集中於日內（2.4 節）；
本研究在日頻上直接測試其中兩項核心構想，**兩者皆未成立**：

- 結構性斷裂標籤的 AUC **0.5286**，低於 SSD 排序本身的 0.5393
- 以標準差計的停損邊界在四條主力臂 × 三個水準上**無一使年化報酬為正**

# 6.2 研究貢獻：實證貢獻五項

**其一，把動作空間設為唯一變因，並完整預先登記。**
既有 DRL 研究各自選定一種動作空間並報告績效，未有研究在同一條管線內
固定其餘一切、僅變動動作空間。假說、效果量門檻、多重比較族與七項核對程序
皆於回測啟動前提交版本庫；偏離逐筆記錄於附錄 D，
**包含一處由研究者自身實作錯誤所致、更正後使核對由失敗轉為通過的情形**。

**其二，分離「動作空間的形狀」與「逐對辨識能力」。**
既有文獻報告學習式交易端優於固定規則時通常未區分兩者，
而兩者的實務意涵相反：前者指向更好的選單設計，後者指向更好的特徵與模型。

## 實證貢獻（續）

**其三，釐清增益的非來源。**
指出損益分解中的「SKIP 貢獻」是**會計恆等式而非技巧歸因**——
底層期望值為負時，隨機跳過任一批配對在期望上都會「避開損失」。
置換檢定證實此點：**兩個配對來源的 SKIP 技巧成分為負**。

**其四，指出文獻成功案例的頻率相依，並提供日頻的受控對照。**
本研究無法分辨「方法有效但日頻訊噪比不足」與「型態出於發表選擇」，
但提供了該型態所缺的那一組對照。

**其五，記錄一組實作錯誤及其對結論的影響。**
修正後主軸 13 格中 **12 格**的績效改善，$\Delta$Sharpe 中位 **+0.236**。

> 一項可陳述的觀察：**以錯誤的臨界值與測錯對象的估計量所篩掉的，
> 系統性地是較好的配對。**

## 6.2.2 關於方法論的一段說明

本研究採取了數項在統計與實驗設計文獻中屬**既有標準**、
但在配對交易比較研究中不常見的作法：

- 以時間取代參數網格為抽樣單位（避免偽重複，Hurlbert, 1984）
- 以信賴區間的寬度區分兩種「不顯著」
- 把單變因原則實作為**程式架構的性質**
- 限制與缺陷的量化揭露

> **本研究不宣稱這些是方法論創見**——它們皆非新方法。
>
> 可陳述者僅為：該領域的慣例與這些標準之間存在落差，
> 而本研究記錄了落差的後果（例如改以時間為抽樣單位後，
> **效果量幾乎不變，但推論的正當性完全不同**）。

# 6.3 對後續研究的建議

| # | 建議 |
| :-- | :--- |
| 一 | **在其他配對來源與格點上重複動作空間消融**——可先以成本較低的 v3（FQI）擴大覆蓋 |
| 二 | **為學習臂固定隨機種子並跑多輪**——同組態兩次執行全距達 2.0–2.8 倍 |
| 三 | **補完門檻水準的掃描**——下界 1.5 從未執行；擴張窗最佳常數多數期選中 3.0，此線索值得追 |
| 四 | **直接驗證槽位週轉假說**——實作「隨機 SKIP 但維持正常槽位配置」的模組重跑；事後重抽無法涵蓋再配置效應 |
| 五 | **若要檢驗形成期的改良，須先解決檢定力問題**——在構造出配對設計之前，擴大母體比擴充特徵維度更具邊際效益 |
| 六 | **重新檢視插補方式對既有研究的影響**——產業中位數插補是隱性管道，影響可達 42.8 pp 的跨產業配對比例差異 |

# 6.3.2 對實務應用的建議

本研究**不建議**將所檢驗的策略直接應用於實務，理由為 5.1 節的三道界線。

> **以成本作為否定理由僅適用於 K-means 那條臂。**
> GICS 來源疊加門檻選擇式後 break-even 達 0.99–1.01%，對 0.58% 有逾七成餘裕。
> **不建議的主要理由在絕對績效與 DSR。**

若仍欲採用配對交易，本研究的證據支持兩項設計取向：

1. **動作空間宜設計為「每期少數次決策」，而非「逐日自由持倉」**
   （證據最強的一項：完整預先登記、3/3 達標、兩種定義下一致），
   且應包含**「拒絕交易」的選項**
2. **不要期待模型能逐對辨識**——把力氣放在選單的設計比放在特徵工程更可能有回報

並須記住第三項：**獲利高度集中於特定市場環境**。
十組配置在平靜期**全部**為負或零 Sharpe，且動盪期報酬的 54%–87%
集中於 2008–2009 兩年。**策略實質上在賣出波動率的均值回歸選擇權。**

# 6.4 結語：過程所揭露的五件事

**其一，表達能力更強的設計可以系統性地更差。**

逐日自由持倉在理論上涵蓋任何固定門檻規則，故若以表達能力為準它應弱優於
門檻選擇。實測相反，且差距不小（換手 2.2–4.1 倍、Sharpe 一律更低）。

> 決策次數愈多，每次決策所依據的訊號愈短、噪音佔比愈高，
> 且每次決策都可能產生一次換手成本——**這個取捨在日頻上是決定性的。**

## 其二，判準必須早於資料，而且要能承受自己的錯誤

本研究把消融的假說、門檻與核對程序在執行前提交版本庫。
其價值在一次具體事件上顯現：

> 分析程式首次執行時核對未通過，而未通過的原因是**核對程式本身**
> 把「左緣」實作錯了。更正後由失敗轉為通過——
> **這正是預先登記最容易被濫用的情境（看到失敗才改程式）。**

本研究的處置是把更正的依據、它只用日期不涉結果這一性質、
以及「它確實是在看到失敗之後才改的」一併寫進偏離紀錄。

> **預先登記的價值不在於不出錯，而在於出錯時留下痕跡。**

## 其三、其四

**其三，「未獲支持」與「無效」不是同一件事，而區分它們需要代價。**
本研究對配對來源之間的比較施加了受控消融，最終仍只能說
「在此設計與樣本下無法被驗證」——因為該設計的最小可偵測效果
大於參照臂自身的績效。

> 把這一節從「結論」降為「描述性觀察」，
> 是本研究**最不情願但最必要**的一個決定。

**其四，設計選擇會以未預期的方式攜帶資訊。**
產業先驗不只存在於明確的 one-hot 編碼，也潛藏於缺失值的插補規則與
距離準則的選擇之中。**研究者可能在自以為移除某項先驗時，實際上仍保留了它。**

## 其五，實作與宣告之間的落差可以大到改變結論的方向

本研究於後期發現多項實作與方法章節所宣告的不符：

- 共整合檢定**誤用了另一個分布的臨界值**
- 均值回歸指標**測量的是錯誤的對象**
- 密度分群的**噪音標記被當成一個群**
- 缺失值的插補規則**悄悄夾帶了產業資訊**
- 對沖權重的**口徑不一致**
- 子視窗回測的**槽位洩漏會讓某一期零配對而不報錯**

> 修正後，交易層的相對效果自 **0/5 顯著變為 3/5 顯著**。
> 這意味著在修正之前，本研究關於交易層的任何結論都建立在一組
> **並非其所描述的管線**之上。
>
> **方法章節寫了什麼，與程式實際做了什麼，是兩件需要分別驗證的事。**

# 本研究最終能夠主張的

> ## 這是一個關於**設計**而非關於**獲利**的結論
>
> 在日頻配對交易上，決定學習式交易端成敗的是**動作空間的形狀**，
> 不是學習演算法的強弱，也不是模型能否辨識每一組配對。
>
> 至於那個形狀所能帶來的增益——
> **它是真的，方向一致，機制可歸因，但它不夠大。**
